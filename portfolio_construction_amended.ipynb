## Research question and evaluation contract

**Question:** For the fixed eight-ETF universe, how do seven stated allocation rules
change net growth, risk, concentration and trading requirements relative to equal
weight, SPY and cash under the same dated information and accounting rules?

This is a retrospective continuation of the 2024 CQF submission. The universe was
chosen with hindsight, the current sample has already been examined, and adjusted
prices and factor files are current vintages. Chronological fitting prevents future
rows entering a fit; it cannot undo universe selection or make this a prospective test.
No strategy is selected as a winner from these tables. A prospective claim would
require freezing the design before new observations arrive and separately validating it.

Primary comparisons are net CAGR and Sharpe relative to equal weight, accompanied
by maximum drawdown, volatility, turnover and costs. Tracking error/information
ratio are relative to SPY; they measure benchmark deviation, not total risk.
Cash is a financing reference. We inspect all strategies, including adverse results;
a higher Sharpe with worse drawdown or excessive turnover is not an unqualified improvement.
No statistical significance or guaranteed return is claimed at this stage.

| Method | Objective / expected-return input | Target constraints |
| --- | --- | --- |
| Equal weight | Fixed 1/N control | Fully invested long-only |
| Minimum variance | Minimize sample annual variance | Sum 1; each weight 0–35% |
| BL maximum Sharpe | Maximize posterior annual excess return / volatility | Sum 1; each weight 0–35% |
| Minimum tracking error | Minimize active-return variance versus SPY | Sum 1; each weight 0–35% |
| Higher moments | Annual mean minus variance penalty plus daily skew reward minus daily excess-kurtosis penalty | Sum 1; each weight 0–35%; multistart feasibility checked |
| Fractional Kelly | Maximize completed-month sample log wealth, then scale by 0.50 | Full Kelly risky sum ≤1; each ≤35%; residual cash |
| BAB | Beta-neutral low/high-beta construction | Gross risky exposure ≤2; residual signed cash; borrow fees |

These are default target limits, not intramonth hard limits: holdings drift until the
next rebalance. BAB's estimated beta neutrality is not realized neutrality. Estimated
expected returns, sample statistics and later realized results are labelled separately.

### Calculation: Input validation and frozen price data

````python
def _numeric_frame(frame, name, minimum_rows=24):
    if not isinstance(frame, pd.DataFrame) or len(frame) < minimum_rows:
        raise ValueError(f'{name} must be a DataFrame with at least {minimum_rows} rows.')
    result = frame.apply(pd.to_numeric, errors='raise').copy()
    if result.index.has_duplicates:
        raise ValueError(f'{name} has duplicate dates.')
    if not np.isfinite(result.to_numpy()).all():
        raise ValueError(f'{name} must contain finite values.')
    return result.sort_index()


def file_sha256(path):
    return hashlib.sha256(Path(path).read_bytes()).hexdigest()


def verify_submission_files(root=None):
    """Verify imported bytes against the checked-in provenance manifest."""
    root = Path(root) if root else DATA_DIR.parent
    manifest = json.loads((root / 'data/manifest.json').read_text(encoding='utf-8'))
    for item in manifest['files']:
        path = root / item['path']
        if not path.is_file() or file_sha256(path) != item['sha256']:
            raise ValueError(f'Submission file missing or changed: {item["path"]}')
    return manifest


def load_submission_prices(path=None, symbols=None, start=None, end=None):
    """Pivot genuine adjusted closes, rejecting duplicate keys and invalid prices.

    Market Cap is a repeated snapshot, not historical market weights. It is never
    used here. A requested end is exclusive. Missing observations stay missing.
    """
    path = Path(path) if path else DATA_DIR / 'efts_dataset.csv'
    raw = pd.read_csv(path)
    required = {'Date', 'ETF', 'Adj Close'}
    if not required.issubset(raw.columns):
        raise ValueError(f'ETF cache requires columns {sorted(required)}.')
    if raw[list(required)].isna().any().any():
        raise ValueError('ETF cache contains missing keys or adjusted prices.')
    raw['Date'] = pd.to_datetime(raw['Date'], errors='raise')
    if raw.duplicated(['Date', 'ETF']).any():
        raise ValueError('ETF cache contains duplicate (Date, ETF) observations.')
    raw['Adj Close'] = pd.to_numeric(raw['Adj Close'], errors='raise')
    if not np.isfinite(raw['Adj Close']).all() or (raw['Adj Close'] <= 0).any():
        raise ValueError('Adjusted prices must be finite and positive.')
    prices = raw.pivot(index='Date', columns='ETF', values='Adj Close').sort_index()
    if symbols is not None:
        missing = set(symbols) - set(prices.columns)
        if missing:
            raise ValueError(f'ETF cache is missing symbols: {sorted(missing)}')
        prices = prices.loc[:, list(symbols)]
    if start is not None:
        prices = prices.loc[prices.index >= pd.Timestamp(start)]
    if end is not None:
        prices = prices.loc[prices.index < pd.Timestamp(end)]
    if prices.empty:
        raise ValueError('No submission prices in the requested date range.')
    prices.columns.name = None
    return prices
````

### Calculation: Snapshot integrity, exchange sessions and realized cash

The complete implementations below fail on missing data or a hash mismatch. No failed download silently becomes synthetic data.

````python
def load_price_snapshot(path, symbols, start, end):
    """Read a provider snapshot only after verifying its immutable manifest."""
    manifest_path = Path(path)
    manifest = json.loads(manifest_path.read_text(encoding='utf-8'))
    price_path = manifest_path.parent / manifest['price_file']
    if hashlib.sha256(price_path.read_bytes()).hexdigest() != manifest['price_sha256']:
        raise ValueError('Price snapshot hash mismatch.')
    actions_path = manifest_path.parent / manifest['actions_file']
    if hashlib.sha256(actions_path.read_bytes()).hexdigest() != manifest['actions_sha256']:
        raise ValueError('Corporate-action snapshot hash mismatch.')
    prices = pd.read_csv(price_path, index_col=0, parse_dates=True, float_precision='round_trip')
    if (list(prices.columns) != manifest['symbols'] or len(prices) != manifest['rows']
            or str(prices.index.min().date()) != manifest['first_price']
            or str(prices.index.max().date()) != manifest['last_price']):
        raise ValueError('Price snapshot schema or coverage differs from its manifest.')
    if (pd.Timestamp(start) < pd.Timestamp(manifest['requested_start'])
            or pd.Timestamp(end) > pd.Timestamp(manifest['requested_end_exclusive'])):
        raise ValueError('Requested dates exceed the frozen snapshot; explicitly refresh Yahoo inputs.')
    if not set(symbols).issubset(prices.columns):
        raise ValueError('Snapshot does not cover the requested symbols.')
    selected = prices.loc[(prices.index >= pd.Timestamp(start)) & (prices.index < pd.Timestamp(end)), symbols]
    return selected, manifest


def audit_prices(prices, start, end, actions=None, calendar='NYSE', jump_threshold=.10):
    """Reject missing sessions/quotes; flag large adjusted moves for review, never repair them."""
    if (not isinstance(prices.index, pd.DatetimeIndex) or prices.index.has_duplicates
            or not prices.index.is_monotonic_increasing or prices.columns.has_duplicates
            or prices.empty or not np.isfinite(prices.to_numpy()).all()
            or (prices <= 0).any().any()):
        raise ValueError('Price audit requires ordered unique dates and finite positive quotes.')
    if calendar:
        import pandas_market_calendars as mcal
        expected = mcal.get_calendar(calendar).valid_days(
            start_date=start, end_date=pd.Timestamp(end) - pd.Timedelta(days=1)).tz_localize(None)
    else:
        expected = pd.bdate_range(start, end, inclusive='left')
    missing, extra = expected.difference(prices.index), prices.index.difference(expected)
    if len(missing) or len(extra):
        raise ValueError(f'Exchange-session mismatch: missing={list(missing.date)}, unexpected={list(extra.date)}')
    returns = prices.pct_change(fill_method=None)
    flags = returns.where(returns.abs() > jump_threshold).stack().rename('Adjusted return').to_frame()
    flags.index.names = ['Date', 'ETF']
    flags['Corporate action'] = ''
    if actions is not None:
        required = {'Date', 'ETF', 'Action', 'Value'}
        if not required.issubset(actions.columns):
            raise ValueError('Corporate actions require Date, ETF, Action, Value.')
        for date, symbol in flags.index:
            related = actions.loc[(pd.to_datetime(actions['Date']) == date) & (actions['ETF'] == symbol)]
            flags.loc[(date, symbol), 'Corporate action'] = '; '.join(
                f'{r.Action}={r.Value}' for r in related.itertuples())
    coverage = pd.DataFrame({'Observations': prices.count(), 'Missing quotes': prices.isna().sum(),
                             'Flagged moves': [(flags.index.get_level_values('ETF') == s).sum() for s in prices]})
    return coverage, flags


def load_sofr_cash_returns(price_dates, manifest_path='data/refresh_20261016/sofr_provenance.json'):
    """Realized ACT/360 overnight proxy, aligned to close-to-close calendar intervals.

    SOFR is observed in arrears and may be revised. It is never a forecast input.
    Simple interest accrues across each quoted overnight period (including weekends).
    The close-to-close application is an explicit accounting approximation.
    """
    manifest_path = Path(manifest_path)
    manifest = json.loads(manifest_path.read_text(encoding='utf-8'))
    path = manifest_path.parent / manifest['file']
    if hashlib.sha256(path.read_bytes()).hexdigest() != manifest['sha256']:
        raise ValueError('SOFR snapshot hash mismatch.')
    data = pd.read_csv(path, na_values=['.'])
    rates = pd.Series(pd.to_numeric(data['SOFR']).to_numpy() / 100,
                      index=pd.to_datetime(data.iloc[:, 0])).dropna().sort_index()
    dates = pd.DatetimeIndex(price_dates)
    if (len(dates) < 2 or dates.has_duplicates or not dates.is_monotonic_increasing
            or rates.index.has_duplicates or not np.isfinite(rates).all()
            or (rates < 0).any() or rates.index.min() > dates.min()
            or (dates.max() - rates.index.max()).days > 4
            or rates.index.to_series().diff().dt.days.max() > 7):
        raise ValueError('SOFR coverage, dates or rates are invalid; no constant-rate fallback.')
    levels = pd.Series(index=pd.date_range(rates.index.min(), dates.max(), freq='D'), dtype=float)
    wealth = 1.0
    for i, (date, rate) in enumerate(rates.items()):
        if date > dates.max():
            break
        stop = min(rates.index[i + 1] if i + 1 < len(rates) else dates.max(), dates.max())
        interval = pd.date_range(date, stop, freq='D')
        values = wealth * (1 + rate * (interval - date).days.to_numpy() / 360)
        levels.loc[interval] = values
        wealth = values[-1]
    selected = levels.reindex(dates)
    if selected.isna().any():
        raise ValueError('SOFR cannot cover every price interval.')
    result = selected.pct_change(fill_method=None).iloc[1:].rename('SOFR realized cash proxy')
    return result, manifest
````

### Calculation: Price acquisition and within-asset returns

````python
def load_prices(mode='synthetic', start='2020-01-01', end='2024-08-23', seed=42,
                cache_path=None):
    """Return a date-by-symbol table and provenance; never silently change mode."""
    start_date, end_date = pd.Timestamp(start), pd.Timestamp(end)
    if pd.isna(start_date) or pd.isna(end_date) or start_date >= end_date:
        raise ValueError('START_DATE must be earlier than END_DATE; use YYYY-MM-DD.')
    symbols = ASSETS + [BENCHMARK]
    if mode in {'synthetic', 'demo'}:
        dates = pd.bdate_range(start_date, end_date, inclusive='left')
        if len(dates) < 253:
            raise ValueError('Choose at least 253 business days for this example.')
        rng = np.random.default_rng(seed)
        market = rng.normal(0.00025, 0.01, len(dates))
        common = rng.normal(0, 0.004, len(dates))
        beta = np.array([1.1, 0.65, 0.95, 0.85, 1.25, 0.9, 1.0, 0.1])
        drift = np.array([.00005, .00001, .00004, .00001, .00008, .00002, .00001, .00003])
        noise = rng.normal(0, 0.006, (len(dates), len(ASSETS)))
        log_returns = market[:, None] * beta + common[:, None] + drift + noise
        levels = np.column_stack([100 * np.exp(np.cumsum(log_returns, axis=0)),
                                  100 * np.exp(np.cumsum(market))])
        prices = pd.DataFrame(levels, index=dates, columns=symbols)
        mode = 'synthetic'
        label = 'SYNTHETIC VALIDATION DATA - not real ETF performance'
        details = {'seed': seed, 'calendar': 'weekdays, including market holidays'}
    elif mode == 'snapshot':
        path = Path(cache_path) if cache_path else Path('data/refresh_20261016/etf_snapshot.json')
        prices, snapshot = load_price_snapshot(path, symbols, start, end)
        label = 'DATED REFRESH - adjusted ETF prices retrieved 16 October 2026'
        details = {'snapshot_manifest': snapshot, 'source_file_sha256': snapshot['price_sha256']}
    elif mode == 'submission':
        path = Path(cache_path) if cache_path else DATA_DIR / 'efts_dataset.csv'
        prices = load_submission_prices(path, symbols, start, end)
        label = 'SUBMISSION SNAPSHOT - adjusted ETF prices supplied with the 2024 study'
        details = {'source_file': path.name, 'source_file_sha256': file_sha256(path),
                   'price_field': 'Adj Close', 'aum_used': False,
                   'vintage': '2024 submission; exact provider download timestamp unknown'}
    elif mode == 'yahoo':
        import yfinance as yf
        try:
            raw = yf.download(symbols, start=start, end=end, auto_adjust=True,
                              progress=False, threads=False, timeout=30,
                              multi_level_index=True)
        except Exception as exc:
            raise RuntimeError('Yahoo download failed. Check your internet connection, '
                               'or explicitly change DATA_MODE to "synthetic".') from exc
        if raw is None or raw.empty or 'Close' not in raw:
            raise ValueError('Yahoo returned no usable prices. Try again later or choose DATA_MODE="synthetic".')
        prices = raw['Close'].copy()
        missing = set(symbols) - set(prices.columns)
        if missing:
            raise ValueError(f'Yahoo did not return these symbols: {sorted(missing)}. No substitute data were used.')
        prices = prices.reindex(columns=symbols)
        prices.index = pd.to_datetime(prices.index).tz_localize(None)
        label = 'YAHOO FINANCE - adjusted historical prices, downloaded now'
        details = {'auto_adjust': True, 'downloaded_utc': datetime.now(timezone.utc).isoformat()}
    else:
        raise ValueError('DATA_MODE must be "snapshot", "submission", "synthetic" or "yahoo".')

    prices = prices.sort_index()
    if prices.index.has_duplicates:
        raise ValueError('Duplicate price dates: review the source before continuing.')
    prices = prices.apply(pd.to_numeric, errors='raise')
    finite_prices = prices.to_numpy()[np.isfinite(prices.to_numpy())]
    if (finite_prices <= 0).any() or np.isinf(prices.to_numpy()).any():
        raise ValueError('Prices must be positive finite numbers or missing observations.')
    if prices.notna().sum().min() < 253:
        raise ValueError('At least one symbol has fewer than 253 prices. Expand the date range.')
    prices.index.name = 'Date'
    provenance = {
        'mode': mode, 'label': label, 'requested_start': start, 'requested_end_exclusive': end,
        'first_price': str(prices.index.min().date()), 'last_price': str(prices.index.max().date()),
        'price_rows': len(prices), 'symbols': symbols,
        'missing_prices_by_symbol': prices.isna().sum().astype(int).to_dict(),
        'prices_csv_sha256': hashlib.sha256(prices.to_csv().encode('utf-8')).hexdigest(),
        **details,
    }
    return prices, provenance


def aligned_returns(prices):
    """Trim unavailable endpoints and reject gaps inside the common price history.

    Dropping an internal return row would erase held assets' gains/losses. This
    function does not invent prices or an exchange calendar for absent dates.
    """
    if not isinstance(prices.index, pd.DatetimeIndex) or prices.index.has_duplicates:
        raise ValueError('Prices require unique datetime observations.')
    prices = prices.sort_index()
    if prices.columns.has_duplicates or prices.empty or prices.isna().all().any():
        raise ValueError('Price columns must be unique and contain observations.')
    start = max(prices.apply(lambda x: x.first_valid_index()))
    end = min(prices.apply(lambda x: x.last_valid_index()))
    common = prices.loc[start:end]
    if common.isna().any().any():
        raise ValueError('Internal missing prices: repair the source or define an explicit valuation policy; return days cannot be dropped.')
    if not np.isfinite(common.to_numpy()).all() or (common <= 0).any().any():
        raise ValueError('Prices must be finite and positive.')
    returns = common.pct_change(fill_method=None).iloc[1:]
    if len(returns) < 252:
        raise ValueError('Fewer than 252 complete aligned return observations. Expand the dates or inspect gaps.')
    if not np.isfinite(returns.to_numpy()).all():
        raise ValueError('Returns contain non-finite values.')
    if (returns <= -1).any().any():
        raise ValueError('A simple return is at or below -100%; inspect the price data.')
    return returns
````

### Calculation: Performance statistics and benchmark comparison

````python
def historical_metrics(series, annual_rf=0.03, periods_per_year=PERIODS_PER_YEAR,
                       cash_returns=None):
    """Annualized metrics from explicit period frequency; default is daily/252."""
    values = np.asarray(series, dtype=float)
    if len(values) < 2 or not np.isfinite(values).all() or (values <= -1).any():
        raise ValueError('At least two finite period returns above -100% are required.')
    if not np.isfinite(annual_rf) or annual_rf <= -1:
        raise ValueError('The annual risk-free assumption must be finite and greater than -100%.')
    if not np.isfinite(periods_per_year) or periods_per_year <= 0:
        raise ValueError('periods_per_year must be positive.')
    volatility = values.std(ddof=1) * np.sqrt(periods_per_year)
    if cash_returns is None:
        cash = np.full(len(values), (1 + annual_rf) ** (1 / periods_per_year) - 1)
    else:
        if isinstance(series, pd.Series) and not cash_returns.index.equals(series.index):
            raise ValueError('Cash returns must match performance dates exactly.')
        cash = np.asarray(cash_returns, dtype=float)
        if cash.shape != values.shape or not np.isfinite(cash).all() or (cash <= -1).any():
            raise ValueError('Invalid performance cash returns.')
    excess = values - cash
    excess_volatility = excess.std(ddof=1) * np.sqrt(periods_per_year)
    downside = np.sqrt(np.mean(np.minimum(excess, 0) ** 2) * periods_per_year)
    wealth = np.concatenate([[1.0], np.cumprod(1 + values)])
    drawdown = wealth / np.maximum.accumulate(wealth) - 1
    return {
        'Annual mean': values.mean() * periods_per_year,
        'CAGR': wealth[-1] ** (periods_per_year / len(values)) - 1,
        'Annual volatility': volatility,
        'Sharpe': excess.mean() * periods_per_year / excess_volatility if excess_volatility > 1e-15 else np.nan,
        'Sortino': excess.mean() * periods_per_year / downside if downside > 0 else np.nan,
        'Max drawdown': drawdown.min(),
    }


def performance_table(returns, annual_rf=0.03, periods_per_year=252, benchmark=None, cash_returns=None):
    data = _numeric_frame(returns, 'returns', minimum_rows=2)
    table = pd.DataFrame({name: historical_metrics(data[name], annual_rf, periods_per_year, cash_returns)
                          for name in data}).T
    if benchmark is not None:
        reference = pd.Series(benchmark).reindex(data.index)
        if not np.isfinite(reference.to_numpy()).all():
            raise ValueError('Benchmark dates must cover returns.')
        active = data.sub(reference, axis=0)
        table['Tracking error'] = active.std(ddof=1) * np.sqrt(periods_per_year)
        table['Information ratio'] = (active.mean() * periods_per_year
                                      / table['Tracking error'].replace(0, np.nan))
    return table
````

### Calculation: CAPM estimation against an aligned market proxy

````python
def capm_diagnostics(returns, benchmark, annual_rf=0.03):
    """OLS on matching daily excess returns; cash rate is an explicit assumption."""
    data = _numeric_frame(returns, 'returns', minimum_rows=24)
    market = pd.Series(benchmark).reindex(data.index)
    if not np.isfinite(market.to_numpy()).all() or market.var() <= 0:
        raise ValueError('Market must be finite, aligned and have positive variance.')
    rf = (1 + annual_rf) ** (1 / 252) - 1
    beta = data.apply(lambda x: x.cov(market) / market.var())
    alpha = (data.mean() - rf) - beta * (market.mean() - rf)
    return pd.DataFrame({'Beta': beta, 'Annual alpha': alpha * 252,
                         'Historical annual excess': (data.mean() - rf) * 252,
                         'CAPM annual excess': beta * (market.mean() - rf) * 252,
                         'R-squared': data.apply(lambda x: x.corr(market) ** 2),
                         'Observations': len(data)})
````

### Calculation: Rolling regressions and training-window selection

````python
def rolling_market_model(returns, benchmark, window=60, annual_rf=0.03):
    """Trailing excess-return OLS coefficients shifted one observation forward."""
    data = _numeric_frame(returns, 'returns', minimum_rows=window + 1)
    market = pd.Series(benchmark).reindex(data.index)
    if not isinstance(window, int) or window < 3 or not np.isfinite(market).all():
        raise ValueError('Window and benchmark are invalid.')
    rf = (1 + annual_rf) ** (1 / 252) - 1
    x, y = market - rf, data - rf
    variance = x.rolling(window).var().replace(0, np.nan)
    beta = y.apply(lambda s: s.rolling(window).cov(x) / variance).shift(1)
    alpha = y.rolling(window).mean().shift(1) - beta.mul(x.rolling(window).mean().shift(1), axis=0)
    # Conditional on the realized market return; this is not an ex-ante return forecast.
    fitted = alpha + beta.mul(x, axis=0)
    return {'beta': beta, 'annual_alpha': alpha * 252,
            'conditional_excess': fitted}


def select_rolling_window(training_returns, benchmark, windows=(30, 60, 90), annual_rf=0.03):
    """Compare conditional return residuals on identical training-only dates."""
    data = _numeric_frame(training_returns, 'training_returns', minimum_rows=max(windows) + 2)
    rf = (1 + annual_rf) ** (1 / 252) - 1
    common = data.index[max(windows):]
    rows = []
    for window in windows:
        fitted = rolling_market_model(data, benchmark, window, annual_rf)['conditional_excess'].loc[common]
        if not np.isfinite(fitted.to_numpy()).all():
            raise ValueError('A candidate rolling regression has a zero-variance market window.')
        errors = data.loc[common] - rf - fitted
        rows.append({'Window': window, 'Conditional return MSE': float(np.mean(errors.to_numpy() ** 2)),
                     'Dates': len(common), 'Assets': data.shape[1],
                     'First scored date': str(common.min().date()),
                     'Last scored date': str(common.max().date())})
    scores = pd.DataFrame(rows).set_index('Window')
    return int(scores['Conditional return MSE'].idxmin()), scores
````

### Calculation: Covariance estimator and dated view register

Sample covariance uses n−1 scaling. Ledoit–Wolf shrinks the maximum-likelihood
covariance toward a scaled identity; its shrinkage coefficient is fitted on each
prior window and recorded. No covariance estimator sees its execution or return date.

The register separates the provider forecast from ETF mapping and model-confidence
assumptions. Publication and mapping-record dates must both precede the decision.
Enabled sets require source, annual arithmetic total-return spread, a 12-month horizon,
activation, exclusive expiry, confidence and its basis. The latest available set
supersedes older sets; expiry does not revive an old forecast. Activation is monthly;
expiry adds a rebalance at the first available close on/after expiry. The prior-only
control uses the identical schedule and costs, including these expiry rebalances.


````python
def estimate_covariance(returns, estimator='sample'):
    """Annual covariance from the supplied prior window; shrinkage uses Ledoit-Wolf.

    The shrinkage estimator uses its ML (n) scaling; sample covariance uses n-1.
    Report that convention rather than silently modifying the library estimator.
    """
    data = _numeric_frame(returns, 'covariance returns', minimum_rows=24)
    if estimator == 'sample':
        covariance, shrinkage = data.cov() * 252, 0.0
    elif estimator == 'ledoit_wolf':
        from sklearn.covariance import LedoitWolf
        fit = LedoitWolf().fit(data.to_numpy())
        covariance = pd.DataFrame(fit.covariance_ * 252, index=data.columns, columns=data.columns)
        shrinkage = float(fit.shrinkage_)
    else:
        raise ValueError('Covariance estimator must be sample or ledoit_wolf.')
    covariance.attrs['shrinkage'] = shrinkage
    return covariance


def validate_view_register(register, symbols):
    """Validate explicitly sourced annual ETF relative-return view sets.

    Disabled qualitative records may have missing numerical/validity fields. Enabled
    sets require a 12-month arithmetic total-return spread and declared model confidence.
    """
    if not isinstance(register, list):
        raise ValueError('View register must be a list of dated view sets.')
    ids, publications = set(), set()
    for record in register:
        if not isinstance(record, dict) or not isinstance(record.get('id'), str) or record['id'] in ids:
            raise ValueError('View-set IDs must be unique strings.')
        ids.add(record['id'])
        if type(record.get('enabled')) is not bool:
            raise ValueError('View-set enabled must be a boolean.')
        for field in ['source_url', 'source_title', 'mapping_assumptions']:
            if not isinstance(record.get(field), str) or not record[field].strip():
                raise ValueError(f'View set needs {field}.')
        date_fields = ['published_on', 'recorded_on'] + (['valid_from', 'expires_on'] if record['enabled'] else [])
        for field in date_fields:
            value = record.get(field)
            try:
                if not isinstance(value, str) or pd.Timestamp(value).strftime('%Y-%m-%d') != value:
                    raise ValueError()
            except Exception as exc:
                raise ValueError(f'View set needs a canonical {field} date.') from exc
        if not record['enabled']:
            if not record.get('inactive_reason'):
                raise ValueError('Disabled view sets need an inactive_reason.')
            continue
        if (record['published_on'] in publications or record.get('horizon_months') != 12
                or record.get('return_definition') != 'annual_arithmetic_total_return_spread'
                or not record['published_on'] <= record['valid_from'] < record['expires_on']
                or record['recorded_on'] > record['valid_from']):
            raise ValueError('Enabled views need unique publication dates, a 12-month spread, and valid recorded/activation/expiry dates.')
        publications.add(record['published_on'])
        if record.get('classification') not in {'provider_forecast_with_mapping', 'research_assumption'}:
            raise ValueError('View classification must distinguish provider mapping from research assumptions.')
        if not isinstance(record.get('views'), list) or not record['views']:
            raise ValueError('Enabled view sets require numerical views.')
        pairs = set()
        for view in record['views']:
            if not isinstance(view, dict):
                raise ValueError('Each numerical view must be an object.')
            pair = (view.get('long'), view.get('short'))
            values = [view.get('annual_spread'), view.get('confidence')]
            if (pair[0] not in symbols or pair[1] not in symbols or pair[0] == pair[1]
                    or frozenset(pair) in pairs
                    or not all(isinstance(v, (float, int)) and np.isfinite(v) for v in values)
                    or not 0 < values[1] < 1 or not view.get('uncertainty_basis')):
                raise ValueError('Views need distinct known assets, finite spreads, confidence strictly between 0 and 1 and an uncertainty basis.')
            pairs.add(frozenset(pair))
    return register


def eligible_view_set(register, decision_close, execution_close):
    """Latest available set supersedes earlier sets; expiry never resurrects an old set.

    Publication/recording are date-only and must both precede the decision close.
    Expiry is exclusive at execution; activation starts at a scheduled rebalance.
    """
    decision, execution = pd.Timestamp(decision_close).normalize(), pd.Timestamp(execution_close).normalize()
    candidates = [r for r in register if r['enabled']
                  and pd.Timestamp(r['published_on']) < decision
                  and pd.Timestamp(r['recorded_on']) < decision
                  and pd.Timestamp(r['valid_from']) <= decision]
    if not candidates:
        return [], None
    newest = max(candidates, key=lambda r: r['published_on'])
    if execution >= pd.Timestamp(newest['expires_on']):
        return [], None
    views = [(v['long'], v['short'], v['annual_spread'], v['confidence']) for v in newest['views']]
    return views, newest['id']
````

### Calculation: Black–Litterman prior, views and posterior

````python
def black_litterman(covariance, prior_weights=None, risk_aversion=3.07, tau=0.1,
                    return_details=False, views=None, omega=None):
    """Annual excess-return posterior, using view-scaled uncertainty.

    A None comparator specifies an absolute annual excess return for one asset.
    Otherwise Q is the annual return difference between two assets.
    """
    names = list(covariance.columns)
    if names != list(covariance.index) or len(names) < 2:
        raise ValueError('Covariance labels and ordering must match.')
    sigma = covariance.to_numpy(dtype=float)
    if not np.isfinite(sigma).all() or not np.allclose(sigma, sigma.T):
        raise ValueError('Covariance must be finite and symmetric.')
    if np.linalg.eigvalsh(sigma).min() < -1e-10:
        raise ValueError('Covariance must be positive semidefinite.')
    if not np.isfinite([tau, risk_aversion]).all() or tau <= 0 or risk_aversion <= 0:
        raise ValueError('tau and risk aversion must be positive.')
    weights = np.full(len(names), 1 / len(names)) if prior_weights is None else np.asarray(prior_weights, dtype=float)
    if weights.shape != (len(names),) or not np.isfinite(weights).all() or (weights < 0).any() or not np.isclose(weights.sum(), 1):
        raise ValueError('Prior weights must be finite, nonnegative and sum to one.')
    # ORIGINAL 2024 SUBMISSION views (21 August 2024); inactive reference only:
    # views = [('GLD', 'MTUM', .15, .90), ('XLK', 'XLE', .08, .70), ('SPHD', 'VUG', .06, .80)]
    views = [] if views is None else list(views)
    for view in views:
        if (len(view) != 4 or view[0] not in names or (view[1] is not None and view[1] not in names)
                or view[0] == view[1] or not np.isfinite(view[2:]).all()
                or not 0 < view[3] <= 1):
            raise ValueError('Views require a known asset, a different known comparator or None for absolute excess, a finite annual Q and confidence in (0, 1].')
    p = np.zeros((len(views), len(names)))
    for j, (long_asset, short_asset, _, _) in enumerate(views):
        p[j, names.index(long_asset)] = 1
        if short_asset is not None:
            p[j, names.index(short_asset)] = -1
    q = np.array([view[2] for view in views])
    confidence = np.array([view[3] for view in views])
    prior = risk_aversion * sigma @ weights
    scaled = tau * sigma
    view_variance = np.diag(p @ scaled @ p.T)
    explicit_omega = omega is not None
    if omega is None:
        omega = np.diag(np.maximum(view_variance, 1e-12) * (1 - confidence) / confidence)
    else:
        omega = np.asarray(omega, dtype=float)
        if (omega.shape != (len(views), len(views)) or not np.isfinite(omega).all()
                or not np.allclose(omega, omega.T)
                or (len(views) and np.linalg.eigvalsh(omega).min() <= 0)):
            raise ValueError('Explicit Omega must be finite, symmetric, positive definite and match views.')
    posterior = prior + scaled @ p.T @ np.linalg.solve(p @ scaled @ p.T + omega, q - p @ prior)
    table = pd.DataFrame(views, columns=['Outperformer', 'Comparator', 'Annual difference', 'Assumed confidence'])
    if explicit_omega:
        table['Assumed confidence'] = np.nan
    if len(views):
        table['Effective marginal weight'] = view_variance / (view_variance + np.diag(omega))
    if return_details:
        mean_covariance = scaled - scaled @ p.T @ np.linalg.solve(p @ scaled @ p.T + omega, p @ scaled)
        labels = [f'{v[0]} - {v[1]}' if v[1] is not None else f'{v[0]} excess' for v in views]
        return {
            'prior': pd.Series(prior, index=names),
            'posterior': pd.Series(posterior, index=names), 'views': table,
            'P': pd.DataFrame(p, index=labels, columns=names),
            'Q': pd.Series(q, index=labels, name='Annual excess/relative view'),
            'Omega': pd.DataFrame(omega, index=labels, columns=labels),
            'mean_covariance': pd.DataFrame(mean_covariance, index=names, columns=names),
            'predictive_covariance': pd.DataFrame(sigma + mean_covariance, index=names, columns=names),
        }
    return pd.Series(prior, index=names), pd.Series(posterior, index=names), table
````

### Calculation: Variance, Sharpe and tracking-error objectives

````python
def optimize_weights(expected_excess, covariance, returns, benchmark, cap=0.35):
    """Long-only, fully invested examples, with checked solver results."""
    names = list(covariance.columns)
    if list(covariance.index) != names or list(expected_excess.index) != names or list(returns.columns) != names:
        raise ValueError('All asset labels must use the same ordering.')
    if not returns.index.equals(benchmark.index):
        raise ValueError('Asset and benchmark dates must match.')
    n = len(names)
    if not 1 / n <= cap <= 1:
        raise ValueError(f'With {n} assets, MAX_WEIGHT must be between {1/n:.3f} and 1.')
    sigma = covariance.to_numpy(dtype=float)
    mu = expected_excess.to_numpy(dtype=float)
    history = returns.to_numpy(dtype=float)
    bench = benchmark.to_numpy(dtype=float)
    if not all(np.isfinite(value).all() for value in [sigma, mu, history, bench]):
        raise ValueError('Optimization inputs must be finite.')
    if (not np.allclose(sigma, sigma.T) or np.linalg.eigvalsh(sigma).min() <= 1e-12):
        raise ValueError('Optimizer covariance must be symmetric positive definite; inspect ill conditioning.')
    equal = np.full(n, 1 / n)

    def variance(w):
        return float(w @ sigma @ w)

    def neg_sharpe(w):
        return float(-(w @ mu) / np.sqrt(max(variance(w), 1e-15)))

    def tracking_variance(w):
        return float(np.var(history @ w - bench, ddof=1) * PERIODS_PER_YEAR)

    result_weights = {'Equal weight': equal}
    for label, objective in [('Minimum variance', variance), ('BL maximum Sharpe', neg_sharpe),
                             ('Minimum tracking error', tracking_variance)]:
        result = minimize(objective, equal, method='SLSQP', bounds=[(0, cap)] * n,
                          constraints=[{'type': 'eq', 'fun': lambda w: w.sum() - 1}],
                          options={'ftol': 1e-12, 'maxiter': 1000})
        w = result.x
        if not result.success or not np.isfinite(w).all() or abs(w.sum() - 1) > 1e-7 or w.min() < -1e-7 or w.max() > cap + 1e-7:
            raise RuntimeError(f'{label} did not find valid weights: {result.message}')
        result_weights[label] = w
    return pd.DataFrame(result_weights, index=names)
````

### Calculation: Monthly compounding and explicit factor-unit conversion

````python
def monthly_compound(daily_returns, trim_boundary_months=True):
    """Compound daily returns; conservatively exclude both boundary months.

    A return-only series cannot prove complete first/last month coverage. Trimming
    prevents partial periods being regressed against full-month factor returns.
    Set False only when complete coverage has been independently established.
    """
    daily = _numeric_frame(daily_returns, 'daily_returns', minimum_rows=40)
    if not isinstance(daily.index, pd.DatetimeIndex):
        raise ValueError('daily_returns must use a DatetimeIndex.')
    periods = daily.index.to_period('M')
    monthly = (1 + daily).groupby(periods).prod() - 1
    if trim_boundary_months:
        monthly = monthly.iloc[1:-1]
    monthly.index = monthly.index.to_timestamp('M')
    return monthly


def load_fama_french_csv(path, unit='percent'):
    """Load a cleaned monthly Fama-French five-factor CSV with explicit units."""
    if unit not in {'percent', 'decimal'}:
        raise ValueError('unit must be "percent" or "decimal".')
    frame = pd.read_csv(Path(path))
    date_column = frame.columns[0]
    dates = frame[date_column].astype(str).str.strip()
    valid = dates.str.fullmatch(r'\d{6}')
    if not valid.all():
        raise ValueError('Cleaned factor file must contain only YYYYMM monthly rows.')
    frame = frame.loc[valid].copy()
    frame.index = pd.to_datetime(dates.loc[valid], format='%Y%m') + pd.offsets.MonthEnd(0)
    required = FACTOR_COLUMNS + ['RF']
    missing = set(required) - set(frame.columns)
    if missing:
        raise ValueError(f'Factor file is missing columns: {sorted(missing)}')
    frame = _numeric_frame(frame[required], 'factor file')
    if unit == 'percent':
        frame /= 100
    if frame.abs().max().max() >= 1:
        raise ValueError('Factor values are implausible decimal returns; check the unit setting.')
    frame.attrs['source_unit'] = unit
    frame.attrs['stored_unit'] = 'decimal'
    return frame
````

### Calculation: Higher-moment, Kelly and BAB allocations

````python
def higher_moment_weights(returns, cap=0.35, risk_aversion=3.0,
                          skew_preference=0.02, kurtosis_penalty=0.005):
    """Composite annual mean/variance and daily standardized-moment preference.

    Deterministic multistart SLSQP checks feasibility, not global optimality.
    Coefficients are explicit heuristic preferences, not a Taylor utility fit.
    """
    data = _numeric_frame(returns, 'returns', minimum_rows=252)
    n = data.shape[1]
    if n < 2:
        raise ValueError("Higher-moment allocation requires at least two assets.")
    if not 1 / n <= cap <= 1:
        raise ValueError(f'cap must be between {1/n:.3f} and 1.')
    if (not np.isfinite([risk_aversion, skew_preference, kurtosis_penalty]).all()
            or min(risk_aversion, skew_preference, kurtosis_penalty) < 0):
        raise ValueError('Higher-moment preferences must be nonnegative.')
    matrix = data.to_numpy()

    def negative_utility(weights):
        series = matrix @ weights
        if series.std(ddof=1) < 1e-12:
            return 1e12
        mean = series.mean() * PERIODS_PER_YEAR
        variance = series.var(ddof=1) * PERIODS_PER_YEAR
        utility = (mean - 0.5 * risk_aversion * variance
                   + skew_preference * skew(series, bias=False)
                   - kurtosis_penalty * kurtosis(series, fisher=True, bias=False))
        return -float(utility)

    equal = np.full(n, 1 / n)
    starts = [equal]
    for index in np.argsort(matrix.mean(axis=0))[-min(3, n):]:
        concentrated = np.full(n, (1 - cap) / (n - 1))
        concentrated[index] = cap
        starts.append(concentrated)
    attempts, candidates = [], []
    for number, start in enumerate(starts):
        result = minimize(negative_utility, start, method='SLSQP', bounds=[(0, cap)] * n,
                          constraints=[{'type': 'eq', 'fun': lambda weights: weights.sum() - 1}],
                          options={'ftol': 1e-12, 'maxiter': 1000})
        w = result.x
        valid = bool(result.success and np.isfinite(w).all() and np.isfinite(result.fun)
                     and result.fun < 1e11 and abs(w.sum() - 1) <= 1e-7
                     and w.min() >= -1e-7 and w.max() <= cap + 1e-7)
        attempts.append({'Start': number, 'Valid': valid, 'Utility': -float(result.fun),
                         'Message': str(result.message)})
        if valid:
            candidates.append(result)
    if not candidates:
        raise RuntimeError(f'Higher-moment optimization found no feasible solution: {attempts}')
    best = min(candidates, key=lambda result: result.fun)
    weights = pd.Series(best.x, index=data.columns, name='Higher-moment utility')
    weights.attrs['solver_attempts'] = attempts
    weights.attrs['local_utility_spread'] = float(max(r.fun for r in candidates) - best.fun)
    return weights


def fractional_kelly_weights(returns, annual_rf=0.03, fraction=0.5, cap=0.35,
                             holding_period='monthly'):
    """Solve capped, unlevered full Kelly with cash, then scale risky weights.

    The fraction is a multiplier of that constrained full-Kelly solution, not
    a required risky exposure. Negative opportunities may leave all wealth cash.
    The cap applies to full Kelly before scaling. Default scenarios are completed
    calendar-month buy-and-hold asset returns (both boundary months excluded).
    Daily mode describes daily rebalancing and is not used by the monthly strategy.
    """
    data = _numeric_frame(returns, 'returns', minimum_rows=252)
    n = data.shape[1]
    if not 0 < fraction <= 1:
        raise ValueError('fraction must be in (0, 1].')
    if not np.isfinite(cap) or not 0 < cap <= 1:
        raise ValueError('cap must be in (0, 1].')
    if not np.isfinite(annual_rf) or annual_rf <= -1:
        raise ValueError('annual_rf must be finite and greater than -1.')
    if holding_period == 'monthly':
        scenarios = monthly_compound(data)
        if len(scenarios) < 10:
            raise ValueError('Monthly Kelly requires at least 10 complete historical months.')
        period_rf = (1 + annual_rf) ** (1 / 12) - 1
    elif holding_period == 'daily':
        scenarios = data
        period_rf = (1 + annual_rf) ** (1 / PERIODS_PER_YEAR) - 1
    else:
        raise ValueError('holding_period must be monthly or daily.')
    matrix = scenarios.to_numpy()

    def negative_log_growth(weights):
        gross = 1 + matrix @ weights + (1 - weights.sum()) * period_rf
        return np.inf if (gross <= 0).any() else -float(np.log(gross).mean())

    start = np.zeros(n)
    result = minimize(negative_log_growth, start, method='SLSQP',
                      bounds=[(0, cap)] * n,
                      constraints=[{'type': 'ineq', 'fun': lambda weights: 1 - weights.sum()}],
                      options={'ftol': 1e-12, 'maxiter': 1000})
    if (not result.success or not np.isfinite(result.x).all()
            or result.x.min() < -1e-7 or result.x.max() > cap + 1e-7
            or result.x.sum() > 1 + 1e-7 or not np.isfinite(result.fun)):
        raise RuntimeError(f'Fractional Kelly optimization failed: {result.message}')
    return pd.Series(result.x * fraction, index=data.columns, name='Fractional Kelly')


def bab_weights(asset_returns, benchmark_returns, gross_limit=2.0):
    """Construct beta-neutral low-minus-high beta weights with bounded gross exposure."""
    assets = _numeric_frame(asset_returns, 'asset_returns', minimum_rows=252)
    benchmark = pd.Series(benchmark_returns, copy=True).reindex(assets.index)
    if benchmark.isna().any() or not np.isfinite(benchmark.to_numpy()).all():
        raise ValueError('benchmark_returns must be finite and aligned with asset_returns.')
    if gross_limit <= 0:
        raise ValueError('gross_limit must be positive.')
    benchmark_variance = benchmark.var(ddof=1)
    if benchmark_variance <= 0:
        raise ValueError('benchmark_returns must have positive variance.')
    betas = assets.apply(lambda column: column.cov(benchmark) / benchmark_variance)
    ordered = betas.sort_values()
    half = len(ordered) // 2
    low_names, high_names = ordered.index[:half], ordered.index[-half:]
    low_beta, high_beta = betas[low_names].mean(), betas[high_names].mean()
    if low_beta <= 0.05 or high_beta <= 0.05:
        raise ValueError('BAB construction requires positive average beta in both legs.')
    raw = pd.Series(0.0, index=assets.columns)
    raw.loc[low_names] = 1 / (len(low_names) * low_beta)
    raw.loc[high_names] = -1 / (len(high_names) * high_beta)
    weights = raw * (gross_limit / raw.abs().sum())
    diagnostics = pd.DataFrame({'Beta': betas, 'Weight': weights,
                                'Leg': np.where(weights > 0, 'Low-beta long', 'High-beta short')})
    diagnostics.attrs['portfolio_beta'] = float(weights @ betas)
    diagnostics.attrs['gross_exposure'] = float(weights.abs().sum())
    return weights.rename('BAB beta-neutral'), diagnostics.sort_values('Beta')
````

### Calculation: Cash, holdings and borrowing ledger

````python
def portfolio_step(weights, cash, asset_returns, daily_rf, daily_borrow_rate, cost=0.0):
    """One self-financing holding day; borrow fees debit cash.

    Rebalance costs reduce initial wealth proportionally before target holdings
    are established. Borrow payments then reduce end-of-day cash and NAV.
    """
    weights = np.asarray(weights, dtype=float)
    asset_returns = np.asarray(asset_returns, dtype=float)
    if (weights.shape != asset_returns.shape or not np.isfinite(weights).all()
            or not np.isfinite(asset_returns).all() or (asset_returns <= -1).any()
            or not np.isfinite([cash, daily_rf, daily_borrow_rate, cost]).all()
            or not 0 <= cost < 1 or daily_borrow_rate < 0
            or not np.isclose(weights.sum() + cash, 1)):
        raise ValueError('Invalid holdings, returns, financing or transaction cost.')
    fee = float(np.maximum(-weights, 0).sum() * daily_borrow_rate)
    closing_assets = weights * (1 + asset_returns)
    closing_cash = cash * (1 + daily_rf) - fee
    gross = float(closing_assets.sum() + closing_cash)
    if gross <= 0:
        raise RuntimeError('Portfolio lost at least 100% after financing and borrow costs.')
    return ((1 - cost) * gross - 1, closing_assets / gross,
            closing_cash / gross, fee)
````

### Calculation: Expanding-window portfolio construction and evaluation

````python
def analyze(prices, annual_rf=0.03, max_weight=0.35, training_days=504,
            transaction_cost_bps=10.0, annual_short_borrow_bps=100.0,
            kelly_fraction=0.5, bab_gross_limit=2.0,
            investor_views=None, views_as_of=None, cash_returns=None,
            view_register=None, apply_registered_views=True,
            training_window_days=None, covariance_estimator='sample',
            evaluation_start_date=None, bab_failure_policy='raise'):
    """Run an expanding-window, monthly walk-forward portfolio comparison.

    For a first earned return ending t, fitting stops at close t-2; execution
    occurs at close t-1. Targets are held between month starts, asset weights drift
    with returns, and one-way turnover is charged at the stated basis-point rate.
    Dated views enter only when the fit cutoff close is strictly after
    their publication date. Date-only sources are treated as available after
    that entire day; the monthly schedule is not changed to force activation.
    """
    if not isinstance(training_days, int) or training_days < 252:
        raise ValueError('TRAINING_DAYS must be an integer of at least 252.')
    if not np.isfinite(transaction_cost_bps) or transaction_cost_bps < 0:
        raise ValueError('TRANSACTION_COST_BPS must be a finite nonnegative number.')
    if not np.isfinite(annual_short_borrow_bps) or annual_short_borrow_bps < 0:
        raise ValueError('ANNUAL_SHORT_BORROW_BPS must be a finite nonnegative number.')

    if training_window_days is not None and (not isinstance(training_window_days, int) or training_window_days < 252):
        raise ValueError('Training window must be None or at least 252 sessions.')
    if bab_failure_policy not in {'raise', 'cash'}:
        raise ValueError('BAB failure policy must be raise or cash.')
    register = validate_view_register([] if view_register is None else view_register, ASSETS)
    if register and investor_views:
        raise ValueError('Use the dated view register or legacy fixed views, not both.')
    configured_views = [] if investor_views is None else list(investor_views)
    publication_date = None
    if configured_views:
        try:
            parsed = datetime.strptime(views_as_of, '%Y-%m-%d')
            if parsed.strftime('%Y-%m-%d') != views_as_of:
                raise ValueError('Noncanonical date')
            publication_date = pd.Timestamp(parsed)
        except (TypeError, ValueError) as exc:
            raise ValueError('Nonempty investor views require VIEWS_AS_OF in YYYY-MM-DD format.') from exc

    returns = aligned_returns(prices)
    if len(returns) <= training_days + 1:
        raise ValueError('The aligned history must contain observations after TRAINING_DAYS.')
    # Validate configured views even if their date lies beyond the sample.
    black_litterman(returns.iloc[:training_days][ASSETS].cov() * PERIODS_PER_YEAR,
                   views=configured_views)
    assets = returns[ASSETS]
    benchmark = returns[BENCHMARK]
    if cash_returns is None:
        cash_series = pd.Series((1 + annual_rf) ** (1 / PERIODS_PER_YEAR) - 1, index=returns.index)
        cash_label = 'constant annual cash-rate assumption'
    else:
        if (not isinstance(cash_returns, pd.Series) or not cash_returns.index.equals(returns.index)
                or not np.isfinite(cash_returns).all() or (cash_returns <= -1).any()):
            raise ValueError('Cash returns must exactly cover all aligned return dates.')
        cash_series = cash_returns.copy()
        cash_label = 'supplied realized cash proxy; fixed annual_rf remains the forecast assumption'
    evaluation = returns.iloc[training_days + 1:]
    if evaluation_start_date is not None:
        evaluation = evaluation.loc[pd.Timestamp(evaluation_start_date):]
    if len(evaluation) < 2:
        raise ValueError('At least two evaluation sessions are required.')
    # Monthly activation, plus execution at the first session on/after view expiry.
    rebalance_dates = evaluation.groupby(evaluation.index.to_period('M')).head(1).index
    expiry_effective_dates = []
    for record in register:
        if record['enabled']:
            execution_positions = np.flatnonzero(returns.index >= pd.Timestamp(record['expires_on']))
            if len(execution_positions) and execution_positions[0] + 1 < len(returns):
                effective = returns.index[execution_positions[0] + 1]
                if effective in evaluation.index:
                    expiry_effective_dates.append(effective)
    rebalance_dates = rebalance_dates.union(pd.DatetimeIndex(expiry_effective_dates)).sort_values()
    positions = [returns.index.get_loc(date) for date in rebalance_dates]
    methods = ['Equal weight', 'Minimum variance', 'BL maximum Sharpe',
               'Minimum tracking error', 'Higher-moment utility',
               'Fractional Kelly', 'BAB beta-neutral']
    current = {name: {'assets': np.zeros(len(ASSETS)), 'cash': 1.0} for name in methods}
    net_returns = {name: [] for name in methods}
    net_dates = []
    target_records = []
    turnover_records = []
    view_records = []
    solver_records = []
    borrow_cost_totals = {name: 0.0 for name in methods}
    last_fit = None
    cost_rate = transaction_cost_bps / 10_000
    daily_rf = (1 + annual_rf) ** (1 / PERIODS_PER_YEAR) - 1
    daily_short_borrow_rate = annual_short_borrow_bps / 10_000 / PERIODS_PER_YEAR
    last_bab_diagnostics = None

    for number, start_position in enumerate(positions):
        stop_position = positions[number + 1] if number + 1 < len(positions) else len(returns)
        # Fit after close t-2, execute at close t-1, earn the return ending t.
        train = returns.iloc[:start_position - 1]
        if training_window_days is not None:
            train = train.iloc[-training_window_days:]
        test = returns.iloc[start_position:stop_position]
        covariance = estimate_covariance(train[ASSETS], covariance_estimator)
        decision_close = train.index[-1]
        eligible = (publication_date is not None
                    and decision_close.normalize() > publication_date)
        active_views = configured_views if eligible else []
        active_set = None
        if register:
            active_views, active_set = eligible_view_set(register, decision_close, returns.index[start_position - 1])
            if not apply_registered_views:
                active_views, active_set = [], None
        prior, posterior, views = black_litterman(covariance, views=active_views)
        view_records.append({'Date': test.index[0], 'Decision close': decision_close,
                             'Execution close': returns.index[start_position - 1],
                             'Publication date': next((r['published_on'] for r in register if r['id'] == active_set), None) if register else (views_as_of if configured_views else None),
                             'Configured views': sum(len(r.get('views', [])) for r in register if r['enabled']) if register else len(configured_views),
                             'Active set': active_set,
                             'Fit start': train.index[0], 'Fit observations': len(train),
                             'Covariance estimator': covariance_estimator,
                             'Shrinkage': covariance.attrs['shrinkage'],
                             'Covariance condition number': float(np.linalg.cond(covariance)),
                             'Active views': len(active_views)})
        targets = optimize_weights(posterior, covariance, train[ASSETS],
                                   train[BENCHMARK], max_weight)
        higher = higher_moment_weights(train[ASSETS], cap=max_weight)
        targets['Higher-moment utility'] = higher
        for attempt in higher.attrs['solver_attempts']:
            solver_records.append({'Date': test.index[0], **attempt,
                                   'Local utility spread': higher.attrs['local_utility_spread']})
        targets['Fractional Kelly'] = fractional_kelly_weights(
            train[ASSETS], annual_rf=annual_rf, fraction=kelly_fraction,
            cap=max_weight)
        try:
            targets['BAB beta-neutral'], last_bab_diagnostics = bab_weights(
                train[ASSETS], train[BENCHMARK], gross_limit=bab_gross_limit)
            view_records[-1]['BAB construction'] = 'beta-neutral target'
        except ValueError as exc:
            if bab_failure_policy != 'cash' or 'positive average beta' not in str(exc):
                raise
            # Explicit experiment policy: do not force leverage through near-zero/negative leg beta.
            targets['BAB beta-neutral'] = pd.Series(0., index=ASSETS)
            betas = train[ASSETS].apply(lambda col: col.cov(train[BENCHMARK]) / train[BENCHMARK].var())
            last_bab_diagnostics = pd.DataFrame({'Beta': betas, 'Weight': 0., 'Leg': 'Cash: ineligible leg beta'})
            last_bab_diagnostics.attrs.update(portfolio_beta=0., gross_exposure=0.)
            view_records[-1]['BAB construction'] = 'cash: nonpositive or near-zero leg beta'
        for name in methods:
            target = targets[name].to_numpy(dtype=float)
            target_cash = 1 - target.sum()
            before = current[name]
            turnover = float((np.abs(target - before['assets']).sum()
                              + abs(target_cash - before['cash'])) / 2)
            turnover_records.append({'Date': test.index[0], 'Portfolio': name,
                                     'One-way turnover': turnover,
                                     'Estimated cost': turnover * cost_rate})
            target_records.append({'Date': test.index[0], 'Portfolio': name,
                                   **dict(zip(ASSETS, target)), 'Cash': target_cash})
            current[name] = {'assets': target.copy(), 'cash': target_cash}

        for row_number, (date, row) in enumerate(test[ASSETS].iterrows()):
            net_dates.append(date)
            asset_returns = row.to_numpy(dtype=float)
            for name in methods:
                weights_now = current[name]['assets']
                cash_now = current[name]['cash']
                cost = turnover_records[-len(methods) + methods.index(name)]['Estimated cost'] if row_number == 0 else 0.0
                net, closing_weights, closing_cash, short_borrow = portfolio_step(
                    weights_now, cash_now, asset_returns, float(cash_series.loc[date]),
                    daily_short_borrow_rate, cost)
                net_returns[name].append(net)
                borrow_cost_totals[name] += short_borrow
                current[name] = {
                    'assets': closing_weights,
                    'cash': closing_cash,
                }
        last_fit = (covariance, prior, posterior, views, targets, last_bab_diagnostics)

    portfolio_returns = pd.DataFrame(net_returns, index=pd.DatetimeIndex(net_dates))
    portfolio_returns[BENCHMARK] = evaluation[BENCHMARK]
    portfolio_returns['Cash proxy'] = cash_series.loc[evaluation.index]
    historical = pd.DataFrame({name: historical_metrics(portfolio_returns[name], annual_rf,
                                                       cash_returns=cash_series.loc[evaluation.index])
                               for name in portfolio_returns}).T
    covariance, prior, posterior, views, asset_weights, bab_diagnostics = last_fit
    cash_weights = 1 - asset_weights.sum(axis=0)
    weights = pd.concat([asset_weights, pd.DataFrame([cash_weights], index=['Cash'])])
    model = pd.DataFrame(index=asset_weights.columns)
    model['Expected annual return'] = [annual_rf + asset_weights[name].to_numpy() @ posterior.to_numpy()
                                       for name in asset_weights]
    model['Annual volatility'] = [np.sqrt(asset_weights[name].to_numpy() @ covariance.to_numpy()
                                          @ asset_weights[name].to_numpy()) for name in asset_weights]
    model['Expected Sharpe'] = ((model['Expected annual return'] - annual_rf)
                                / model['Annual volatility'])
    target_history = pd.DataFrame(target_records).set_index(['Date', 'Portfolio'])
    turnover = pd.DataFrame(turnover_records).set_index(['Date', 'Portfolio'])
    years = len(portfolio_returns) / PERIODS_PER_YEAR
    trading = turnover.groupby('Portfolio').sum()
    trading['Estimated short-borrow cost'] = pd.Series(borrow_cost_totals)
    trading['Estimated total implementation cost'] = (trading['Estimated cost']
                                                       + trading['Estimated short-borrow cost'])
    trading['Annualized one-way turnover'] = trading['One-way turnover'] / years
    trading['Cost assumption (bps)'] = transaction_cost_bps
    trading = trading.reindex(methods)
    return dict(returns=returns, covariance=covariance, prior=prior, posterior=posterior,
                views=views, weights=weights, bab_diagnostics=bab_diagnostics,
                weight_history=target_history,
                turnover=turnover, trading=trading, historical=historical, model=model,
                portfolio_returns=portfolio_returns, evaluation_start=evaluation.index[0],
                evaluation_end=evaluation.index[-1], training_days=training_days,
                transaction_cost_bps=transaction_cost_bps,
                annual_short_borrow_bps=annual_short_borrow_bps,
                kelly_fraction=kelly_fraction, bab_gross_limit=bab_gross_limit,
                views_as_of=view_records[-1]['Publication date'] if register else (views_as_of if configured_views else None),
                configured_view_count=sum(len(r.get('views', [])) for r in register if r['enabled']) if register else len(configured_views),
                view_history=pd.DataFrame(view_records).set_index('Date'),
                solver_history=pd.DataFrame(solver_records).set_index(['Date', 'Start']),
                cash_returns=cash_series, cash_label=cash_label,
                view_register=register, max_weight=max_weight, bab_failure_policy=bab_failure_policy, training_window_days=training_window_days,
                covariance_estimator=covariance_estimator, apply_registered_views=apply_registered_views)
````

### Calculation: Development selection, paired uncertainty and diagnostics

The functions below run the fixed experiment registry, paired circular block bootstrap,
financing grid, drawdown duration, exposure and risk-contribution calculations. Every
function is visible here and tested against its callable Python counterpart.


````python
def run_research_experiments(prices, cash_returns=None, view_register=None,
                             development_start='2022-01-04', development_end='2023-12-31',
                             evaluation_start='2024-01-01', annual_rf=.03,
                             max_weight=.35, transaction_cost_bps=10,
                             annual_short_borrow_bps=100, training_days=504,
                             kelly_fraction=.5, bab_gross_limit=2.):
    """Select one configuration using development minimum-variance volatility only.

    Candidate windows/estimators are fixed before this run, but the entire historical
    sample has already been inspected. Evaluation never participates in selection.
    Every evaluation starts from cash on a common date and pays its initial turnover.
    """
    start, end, later = map(pd.Timestamp, [development_start, development_end, evaluation_start])
    if not start < end < later or prices.index.max() < later:
        raise ValueError('Need disjoint chronological development and evaluation windows.')
    development_prices = prices.loc[:end]
    development_cash = None if cash_returns is None else cash_returns.loc[development_prices.index[1:]]
    candidates = [('expanding_sample', None, 'sample'), ('252_sample', 252, 'sample'),
                  ('504_sample', 504, 'sample'), ('504_ledoit_wolf', 504, 'ledoit_wolf')]
    common = dict(annual_rf=annual_rf, max_weight=max_weight, training_days=training_days,
                  transaction_cost_bps=transaction_cost_bps,
                  annual_short_borrow_bps=annual_short_borrow_bps,
                  kelly_fraction=kelly_fraction, bab_gross_limit=bab_gross_limit,
                  view_register=[] if view_register is None else view_register, bab_failure_policy='cash')
    development, evaluations, rows = {}, {}, []
    dates = None
    for label, window, estimator in candidates:
        result = analyze(development_prices, cash_returns=development_cash,
                         training_window_days=window, covariance_estimator=estimator,
                         evaluation_start_date=start, **common)
        if dates is not None and not dates.equals(result['portfolio_returns'].index):
            raise ValueError('Candidate development dates differ.')
        dates = result['portfolio_returns'].index
        development[label] = result
        metric = result['historical'].loc['Minimum variance']
        rows.append({'Configuration': label, 'Window': 'expanding' if window is None else str(window),
                     'Estimator': estimator, 'Development first': str(dates[0].date()),
                     'Development last': str(dates[-1].date()), 'Development sessions': len(dates),
                     'Selection score: minimum-variance volatility': metric['Annual volatility'],
                     'Development CAGR': metric['CAGR'], 'Development Sharpe': metric['Sharpe']})
    registry = pd.DataFrame(rows).set_index('Configuration')
    # Stable tie break by declared configuration label, never by evaluation outcome.
    selected = registry.sort_values(['Selection score: minimum-variance volatility', 'Configuration']).index[0]
    registry['Selected on development'] = registry.index == selected
    for label, window, estimator in candidates:
        evaluations[label] = analyze(prices, cash_returns=cash_returns,
                                    training_window_days=window, covariance_estimator=estimator,
                                    evaluation_start_date=later, **common)
    selected_result = evaluations[selected]
    eval_dates = selected_result['portfolio_returns'].index
    if any(not r['portfolio_returns'].index.equals(eval_dates) for r in evaluations.values()):
        raise ValueError('Candidate evaluation dates differ.')
    comparison = pd.concat({label: r['historical'] for label, r in evaluations.items()}, names=['Configuration', 'Portfolio'])
    # Same schedule, covariance, constraints and costs isolate the registered views.
    prior_result = analyze(prices, cash_returns=cash_returns,
                           training_window_days=selected_result['training_window_days'],
                           covariance_estimator=selected_result['covariance_estimator'],
                           evaluation_start_date=later, apply_registered_views=False, **common)
    if not prior_result['weight_history'].index.equals(selected_result['weight_history'].index):
        raise ValueError('Prior and views comparison must use identical execution schedules.')
    return {'registry': registry, 'selected': selected, 'result': selected_result,
            'development': development, 'evaluations': evaluations,
            'evaluation_comparison': comparison, 'prior_result': prior_result,
            'development_end': str(end.date()), 'evaluation_start': str(later.date())}


def paired_block_bootstrap(returns, cash_returns, benchmark='Equal weight',
                           block_length=21, replications=2000, seed=42, confidence=.95):
    """Paired circular moving-block percentile intervals for CAGR/Sharpe differences.

    Resample identical contiguous indices for every strategy and cash. Preserve
    within-block dependence and cross-strategy pairing. Targets are not refitted;
    intervals are conditional on the realized strategy paths and stationarity.
    Pointwise intervals are unadjusted for the many research comparisons.
    """
    data = _numeric_frame(returns, 'bootstrap returns', minimum_rows=60)
    if (benchmark not in data or not cash_returns.index.equals(data.index)
            or not np.isfinite(cash_returns).all() or (data <= -1).any().any()
            or (cash_returns <= -1).any() or not isinstance(block_length, int)
            or not 1 <= block_length <= len(data) // 2
            or not isinstance(replications, int) or replications < 200
            or not 0 < confidence < 1):
        raise ValueError('Invalid bootstrap inputs, pairing, block length or replication count.')
    names, values, cash = list(data), data.to_numpy(), cash_returns.to_numpy()
    n, columns = values.shape
    base = names.index(benchmark)

    def statistics(v, c):
        growth = np.expm1(np.log1p(v).mean(axis=0) * 252)
        excess = v - c[:, None]
        vol = excess.std(axis=0, ddof=1)
        sharpe = np.divide(excess.mean(axis=0) * np.sqrt(252), vol,
                           out=np.full(columns, np.nan), where=vol > 1e-15)
        return np.stack([growth - growth[base], sharpe - sharpe[base]], axis=1)

    rng = np.random.default_rng(seed)
    samples = np.empty((replications, columns, 2))
    blocks = int(np.ceil(n / block_length))
    for b in range(replications):
        starts = rng.integers(0, n, size=blocks)
        indices = ((starts[:, None] + np.arange(block_length)) % n).ravel()[:n]
        samples[b] = statistics(values[indices], cash[indices])
    point = statistics(values, cash)
    tail = (1 - confidence) / 2
    rows = []
    for j, name in enumerate(names):
        if name == benchmark:
            continue
        for k, metric in enumerate(['CAGR difference', 'Sharpe difference']):
            draws = samples[:, j, k]
            finite = draws[np.isfinite(draws)]
            lower, upper = np.quantile(finite, [tail, 1 - tail]) if len(finite) else [np.nan, np.nan]
            rows.append({'Portfolio': name, 'Metric': metric, 'Estimate': point[j, k],
                         'Lower': lower, 'Upper': upper, 'Block sessions': block_length,
                         'Replications': replications, 'Finite replicates': len(finite),
                         'Benchmark': benchmark, 'Confidence': confidence,
                         'Inference': 'pointwise, conditional, exploratory'})
    return pd.DataFrame(rows).set_index(['Portfolio', 'Metric'])


def drawdown_duration(returns):
    """Longest/current underwater run in observed sessions, including initial NAV=1."""
    values = np.asarray(returns, dtype=float)
    if not len(values) or not np.isfinite(values).all() or (values <= -1).any():
        raise ValueError('Drawdown needs finite returns above -100%.')
    wealth = np.r_[1., np.cumprod(1 + values)]
    drawdown = wealth / np.maximum.accumulate(wealth) - 1
    longest, current = 0, 0
    for value in drawdown[1:]:
        current = current + 1 if value < -1e-12 else 0
        longest = max(longest, current)
    return {'Longest underwater sessions': longest, 'Current underwater sessions': current}


def portfolio_risk_diagnostics(result):
    """Exposure, concentration, signed risk contribution and realized BAB market beta."""
    history = result['weight_history']
    risky = history[ASSETS]
    gross = risky.abs().sum(axis=1)
    target_exposure = pd.DataFrame({'Target gross risky': gross, 'Target net risky': risky.sum(axis=1),
                                    'Target cash': history['Cash'],
                                    'Target max absolute weight': risky.abs().max(axis=1),
                                    'Risky HHI (absolute normalized)': risky.abs().div(gross.replace(0, np.nan), axis=0).pow(2).sum(axis=1)})
    # HHI is undefined for an all-cash portfolio, not zero concentration.
    target_exposure.loc[gross < 1e-12, 'Risky HHI (absolute normalized)'] = np.nan
    summary = target_exposure.groupby('Portfolio').mean()
    summary = summary.join(result['trading'][['Annualized one-way turnover', 'Estimated total implementation cost']])
    durations = pd.DataFrame({name: drawdown_duration(result['portfolio_returns'][name]) for name in result['weights']}).T
    summary = summary.join(durations)
    sigma = result['covariance'].to_numpy()
    contributions = {}
    for name in result['weights']:
        w = result['weights'].loc[ASSETS, name].to_numpy()
        variance = w @ sigma @ w
        contributions[name] = w * (sigma @ w) / variance if variance > 1e-15 else np.full(len(w), np.nan)
    risk_share = pd.DataFrame(contributions, index=ASSETS)
    excess = result['portfolio_returns'].sub(result['cash_returns'].loc[result['portfolio_returns'].index], axis=0)
    beta = excess['BAB beta-neutral'].rolling(63).cov(excess[BENCHMARK]) / excess[BENCHMARK].rolling(63).var()
    return {'summary': summary, 'target_exposure': target_exposure,
            'latest_signed_variance_contributions': risk_share,
            'realized_bab_beta': beta.rename('63-session realized excess-return beta').to_frame()}


def financing_sensitivity(result, transaction_costs=(0, 10, 25), borrow_costs=(0, 100, 300), annual_rf=.03):
    """Replay fixed target paths on a declared cost grid; no cost-aware refitting."""
    dates = result['portfolio_returns'].index
    cash = result['cash_returns'].loc[dates]
    rows, paths = [], {}
    for cost in transaction_costs:
        for borrow in borrow_costs:
            for name in result['weights']:
                target = result['weight_history'].xs(name, level='Portfolio')
                series, _ = evaluate_targets(result['returns'].loc[dates, ASSETS], target,
                                             annual_rf, cost, borrow, cash_returns=cash)
                metrics = historical_metrics(series, annual_rf, cash_returns=cash)
                rows.append({'Transaction bps': cost, 'Borrow bps': borrow, 'Portfolio': name, **metrics})
                paths[(cost, borrow, name)] = series
    return pd.DataFrame(rows).set_index(['Transaction bps', 'Borrow bps', 'Portfolio']), paths


def view_confidence_sensitivity(prices, selected_result, annual_rf=.03, confidence_levels=(.25, .5, .75)):
    """Change model confidence only for explicitly enabled sets; never invent views."""
    import copy
    register = selected_result['view_register']
    columns = ['Confidence', 'Portfolio', 'Status', 'CAGR', 'Sharpe', 'Max drawdown']
    if not any(r['enabled'] for r in register):
        return pd.DataFrame([{'Confidence': c, 'Portfolio': 'BL maximum Sharpe',
                              'Status': 'Not estimable: no enabled numerical view sets'} for c in confidence_levels], columns=columns)
    rows = []
    for confidence in confidence_levels:
        modified = copy.deepcopy(register)
        for record in modified:
            if record['enabled']:
                for view in record['views']:
                    view['confidence'] = confidence
                    view['uncertainty_basis'] = 'Research sensitivity override; not provider confidence.'
        run = analyze(prices, annual_rf=annual_rf, max_weight=selected_result['max_weight'],
                      training_days=selected_result['training_days'],
                      kelly_fraction=selected_result['kelly_fraction'], bab_gross_limit=selected_result['bab_gross_limit'],
                      transaction_cost_bps=selected_result['transaction_cost_bps'],
                      annual_short_borrow_bps=selected_result['annual_short_borrow_bps'],
                      cash_returns=selected_result['cash_returns'], view_register=modified,
                      training_window_days=selected_result['training_window_days'],
                      covariance_estimator=selected_result['covariance_estimator'],
                      evaluation_start_date=selected_result['evaluation_start'],
                      bab_failure_policy=selected_result['bab_failure_policy'])
        rows.append({'Confidence': confidence, 'Portfolio': 'BL maximum Sharpe',
                     'Status': 'Evaluated' if run['view_history']['Active views'].gt(0).any() else 'No eligible view in evaluated dates',
                     **run['historical'].loc['BL maximum Sharpe'].to_dict()})
    return pd.DataFrame(rows)


def research_summary(experiment, intervals, risk):
    """One source of headline numbers: evaluation metrics, intervals and exposures."""
    result = experiment['result']
    table = result['historical'].copy()
    table['CAGR minus equal weight'] = table['CAGR'] - table.loc['Equal weight', 'CAGR']
    table['Sharpe minus equal weight'] = table['Sharpe'] - table.loc['Equal weight', 'Sharpe']
    for metric, prefix in [('CAGR difference', 'CAGR delta'), ('Sharpe difference', 'Sharpe delta')]:
        subset = intervals.xs(metric, level='Metric')
        table[f'{prefix} lower'] = subset['Lower']
        table[f'{prefix} upper'] = subset['Upper']
    table = table.join(risk['summary'])
    table['Configuration'] = experiment['selected']
    table['Interpretation'] = 'Retrospective; conditional pointwise intervals; no prospective claim'
    return table
````

## 4. Fit and evaluate the portfolio methods

All strategies follow the same chronological contract. After the initial training history, each monthly target uses only observations strictly before its rebalance date. Asset weights drift between rebalances. The simulation records one-way turnover and deducts the stated trading costs; short portfolios also incur the stated borrow cost.

| Portfolio | Maintained objective |
| --- | --- |
| Equal weight | Equal long-only asset targets |
| Minimum variance | Minimum annualized covariance-based variance |
| BL maximum Sharpe | Maximum expected Sharpe using the Black–Litterman posterior |
| Minimum tracking error | Minimum annualized active-return variance versus SPY |
| Higher-moment utility | Explicit annual mean/variance utility with sample skew reward and excess-kurtosis penalty |
| Fractional Kelly | Constrained full log-growth optimum, scaled by the Kelly fraction, plus cash |
| BAB beta-neutral | Low-beta long/high-beta short legs scaled to zero estimated market beta and bounded gross exposure |

Covariance and correlation remain separate. Expected returns and covariance use annual units; daily performance statistics use the documented 252-observation convention.


### Information and execution timing

For a monthly target whose first earned return ends on session $t$:

| Event | Latest allowed information / accounting time |
| --- | --- |
| Fit and decide | After the close at $t-2$ |
| Execute target | Close at $t-1$, with stated turnover cost |
| Earn first return | Close $t-1$ to close $t$ |
| Dated views | Publication date strictly earlier than the fit-cutoff date |

This leaves a full session to submit a close order. The price is an adjusted-close
execution proxy; volume, auction fills and market impact are not modeled. The first
evaluation month may be partial. Thereafter targets become effective on each month's
first session. Costs reduce NAV before the holding interval. Signed risky weights plus
cash sum to one; cash receives financing and pays short borrow, and all holdings are
renormalized by end-of-interval NAV. Turnover is half the absolute change in risky **and
cash** weights, explicitly a model convention rather than a broker fee schedule.


The bounded window/estimator experiment applies one explicit BAB eligibility rule:
if either beta-sorted leg has mean beta ≤0.05, hold cash until a later scheduled fit
can construct valid legs. This avoids dividing by near-zero or negative beta.
Every cash fallback is reported; it is not a beta-neutral trade or an omitted loss.
All configurations use this rule. Other solver errors stop the run. The lower-level
`analyze` API defaults to raising on an infeasible BAB construction unless this policy
is explicitly requested.

### Black–Litterman matrices and return-vector comparison

These tables restore the original explicit P, Q, Omega and posterior-covariance
calculations. All use the **last rebalance's prior window**, whose cutoff is shown.
With no active views, the posterior remains the prior and the view tables are empty. When views are supplied, P maps asset returns into relative views; Q is an annual excess-return difference;
Omega scales view uncertainty from the same covariance. The covariance of the
posterior mean and the predictive covariance `Sigma + mean covariance` are different
objects. The main strategies continue to use estimated return covariance `Sigma`;
the predictive-covariance allocation below is a labelled sensitivity.

Historical means, SPY/ACWI CAPM, equilibrium prior and BL posterior are compared
as annual **excess** returns. Summary averages are explicitly equal weighted.

All means here are **annual excess returns**. The prior is $\pi=\delta\Sigma w_0$,
with equal prior weights (a modeling choice, not market capitalization equilibrium).
The posterior is $\pi+\tau\Sigma P^T(P\tau\Sigma P^T+\Omega)^{-1}(Q-P\pi)$.
Confidence $c$ maps to diagonal view variance $(1-c)/c$ times the corresponding
$P\tau\Sigma P^T$ diagonal. This is an explicit uncertainty convention; confidence
percentages are not empirically calibrated probabilities. No numerical views are active.
Posterior mean uncertainty is distinct from asset return covariance; predictive covariance
adds that uncertainty to $\Sigma$. Main optimizers use the stated sample return covariance.

### Calculation: Mean–variance allocation with an explicit cash balance

````python
def mean_variance_weights(expected_excess, covariance, risk_aversion=3.07, cap=0.35,
                          allow_cash=True):
    """Optimize annual excess utility mu'w - lambda/2 w'Sigma w."""
    mu = pd.Series(expected_excess, dtype=float)
    sigma = pd.DataFrame(covariance).loc[mu.index, mu.index].to_numpy(dtype=float)
    n = len(mu)
    if (not np.isfinite(mu).all() or not np.isfinite(sigma).all()
            or not np.allclose(sigma, sigma.T) or np.linalg.eigvalsh(sigma).min() < -1e-10
            or not np.isfinite(risk_aversion) or risk_aversion <= 0
            or not 0 < cap <= 1 or (not allow_cash and cap * n < 1)):
        raise ValueError('Invalid moments, risk aversion or allocation cap.')
    budget = {'type': 'ineq' if allow_cash else 'eq', 'fun': lambda w: 1 - w.sum()}
    start = np.zeros(n) if allow_cash else np.full(n, 1 / n)
    result = minimize(lambda w: float(-mu.to_numpy() @ w + .5 * risk_aversion * w @ sigma @ w),
                      start, method='SLSQP', bounds=[(0, cap)] * n, constraints=[budget],
                      options={'ftol': 1e-12, 'maxiter': 1000})
    w = result.x
    if (not result.success or not np.isfinite(w).all() or w.min() < -1e-7
            or w.max() > cap + 1e-7 or w.sum() > 1 + 1e-7
            or (not allow_cash and abs(w.sum() - 1) > 1e-7)):
        raise RuntimeError(f'Mean-variance optimization failed: {result.message}')
    return pd.Series(w, index=mu.index)
````

### Calculation: Third/fourth-order moments and portfolio contractions

````python
def higher_moment_diagnostics(returns, weights):
    """Population central tensors contracted into the actual portfolio moments.

    Raw asset moments must be contracted before standardizing by portfolio
    volatility. Contracting standardized asset co-moments with raw weights does
    not produce portfolio skewness/kurtosis. No small-sample correction is used.
    """
    data = _numeric_frame(returns, 'returns')
    w = pd.Series(weights).reindex(data.columns).to_numpy(dtype=float)
    if not np.isfinite(w).all():
        raise ValueError('Weights must cover each asset.')
    x = data.to_numpy() - data.mean().to_numpy()
    n = len(x)
    second = x.T @ x / n
    third = np.einsum('ti,tj,tk->ijk', x, x, x, optimize=True) / n
    fourth = np.einsum('ti,tj,tk,tl->ijkl', x, x, x, x, optimize=True) / n
    variance = float(w @ second @ w)
    if variance <= 0 or (np.diag(second) <= 0).any():
        raise ValueError('Higher moments require positive portfolio and asset variances.')
    skewness = float(np.einsum('i,j,k,ijk', w, w, w, third) / variance ** 1.5)
    kurtosis = float(np.einsum('i,j,k,l,ijkl', w, w, w, w, fourth) / variance ** 2)
    z = x / np.sqrt(np.diag(second))
    pair_skew = (z ** 2).T @ z / n
    pair_kurt = (z ** 2).T @ (z ** 2) / n
    portfolio = x @ w
    checks = pd.DataFrame({
        'Tensor contraction': [variance, skewness, kurtosis - 3],
        'Direct portfolio sample': [np.mean(portfolio ** 2),
                                    np.mean(portfolio ** 3) / variance ** 1.5,
                                    np.mean(portfolio ** 4) / variance ** 2 - 3]},
        index=['Daily variance', 'Skewness', 'Excess kurtosis'])
    checks['Absolute difference'] = (checks.iloc[:, 0] - checks.iloc[:, 1]).abs()
    return {'third': third, 'fourth': fourth, 'checks': checks,
            'pair_coskewness': pd.DataFrame(pair_skew, index=data.columns, columns=data.columns),
            'pair_cokurtosis': pd.DataFrame(pair_kurt, index=data.columns, columns=data.columns)}
````

### Co-skewness, co-kurtosis and portfolio-moment verification

The original third/fourth-order tensors are restored using **raw daily central
moments**. Contracting them with the higher-moment optimizer's own weights and then
standardizing must reproduce skewness and excess kurtosis of that portfolio's
sample return series. Population moments are used consistently for this identity;
the optimizer's finite-sample corrected statistics have a different convention.

The displayed pairwise slices are `E[z_i^2 z_j]` and `E[z_i^2 z_j^2]`, where z is an
asset's standardized daily return. The latter is raw co-kurtosis; subtracting 3
from every off-diagonal entry would not define a valid general excess co-moment.

The allocation utility is $252\bar r-3(252s^2)/2+0.02\,skew(r)-0.005\,excessKurt(r)$.
Annual mean/variance and daily standardized moments have explicit preference coefficients.
This is a composite heuristic, not a validated expected-utility Taylor expansion.
Four deterministic starts report feasibility and local utility spread; agreement does not
prove a global optimum. Portfolio moments always use this optimizer's own weights.


````python
moment_diagnostics = higher_moment_diagnostics(last_train[ASSETS], result["weights"].loc[ASSETS, "Higher-moment utility"])
display(moment_diagnostics["checks"])
assert moment_diagnostics["checks"]["Absolute difference"].max() < 1e-8
print("Tensor shapes:", moment_diagnostics["third"].shape, moment_diagnostics["fourth"].shape)
diagnostic_tables.update(higher_moment_checks=moment_diagnostics["checks"],
                         pair_coskewness=moment_diagnostics["pair_coskewness"],
                         pair_cokurtosis=moment_diagnostics["pair_cokurtosis"])
show_heatmap(moment_diagnostics["pair_coskewness"], "Pairwise co-skewness E[z_i squared z_j]", "pair_coskewness", centered=True)
show_heatmap(moment_diagnostics["pair_cokurtosis"], "Raw pairwise co-kurtosis E[z_i squared z_j squared]", "pair_cokurtosis")
````

### Calculation: Evaluate a dated target history using the holdings ledger

````python
def evaluate_targets(asset_returns, targets, annual_rf=.03, transaction_cost_bps=10,
                     annual_short_borrow_bps=100, cash_returns=None):
    """Evaluate precomputed dated targets with the same ledger as the main run.

    Caller is responsible for fitting targets before their effective dates.
    A target table includes asset columns and Cash and must start on day one.
    """
    returns = _numeric_frame(asset_returns, 'asset_returns', minimum_rows=2)
    targets = _numeric_frame(targets, 'targets', minimum_rows=1)
    names = list(returns.columns)
    if (list(targets.columns) != names + ['Cash'] or targets.index[0] != returns.index[0]
            or not targets.index.isin(returns.index).all()
            or not np.allclose(targets.sum(axis=1), 1)):
        raise ValueError('Targets must cover the initial date, assets and cash budget.')
    if (not np.isfinite([annual_rf, transaction_cost_bps, annual_short_borrow_bps]).all()
            or annual_rf <= -1 or min(transaction_cost_bps, annual_short_borrow_bps) < 0):
        raise ValueError('Invalid financing/cost assumptions.')
    rf = (1 + annual_rf) ** (1 / 252) - 1
    if cash_returns is not None and (not cash_returns.index.equals(returns.index)
                                    or not np.isfinite(cash_returns).all()
                                    or (cash_returns <= -1).any()):
        raise ValueError('Cash returns must exactly match evaluation dates.')
    borrow = annual_short_borrow_bps / 10000 / 252
    w, cash = np.zeros(len(names)), 1.0
    result, trades = [], []
    for date, row in returns.iterrows():
        cost = 0.0
        if date in targets.index:
            target = targets.loc[date, names].to_numpy()
            target_cash = float(targets.loc[date, 'Cash'])
            turnover = (np.abs(target - w).sum() + abs(target_cash - cash)) / 2
            cost = turnover * transaction_cost_bps / 10000
            trades.append({'Date': date, 'One-way turnover': turnover, 'Estimated cost': cost})
            w, cash = target, target_cash
        period_cash = rf if cash_returns is None else float(cash_returns.loc[date])
        net, w, cash, _ = portfolio_step(w, cash, row.to_numpy(), period_cash, borrow, cost)
        result.append(net)
    return pd.Series(result, index=returns.index), pd.DataFrame(trades).set_index('Date')
````

No view remains with fewer than 30 calendar days before quarter end. Display effective Bayesian weight rather than calling an assumed confidence a provider probability.

**Allocation.** Maximize annual mean/variance utility less one-way transaction costs scaled by a 21-session holding horizon. Risk aversion is 3.07; announced targets are long-only, fully invested, capped at 35%, and at most 15 percentage points from equal prior weights (effective maximum 27.5%). At execution, limit actual one-way turnover including cash to 10%; initial entry from all cash is exempt. Partial execution and market drift can leave holdings outside announced-target bounds, which are reported separately. No Q refit uses the execution-day return.

**Controls.** All five paths share dates and costs: the JPM scenario, a matched prior, a fixed five-percentage-point GLD overweight, a GLD-exposure-matched equal-weight remainder, and equal weight scaled with cash toward the scenario's ex ante risk. Controls do not use later realized returns. The cash-only risk control cannot lever up; turnover restrictions can also prevent exact matching. The audit reports actual risk and GLD exposure gaps.

**Evidence.** Rules were developed on 9 October after inspecting these prices. The July-publication replay is a retrospective reconstruction, first affecting the return ending 3 August. Main results retain strict recorded-date eligibility. The revised design is separately frozen for subsequent observations; the short historical sample and unfinished quarter cannot establish alpha or forecast success.


````python
def load_gold_valuations(forecast):
    """Read the pinned issuer NAV/ounces pairs; never silently refresh them."""
    path = Path(forecast['valuation_file'])
    if hashlib.sha256(path.read_bytes()).hexdigest() != forecast['valuation_sha256']:
        raise ValueError('Gold valuation file hash mismatch.')
    frame = pd.read_csv(path, index_col='Date', parse_dates=True)
    if (list(frame.columns) != ['Ounces per share', 'NAV per share']
            or frame.index.has_duplicates or not frame.index.is_monotonic_increasing
            or not np.isfinite(frame.to_numpy()).all() or (frame <= 0).any().any()):
        raise ValueError('Invalid dated issuer valuation pairs.')
    return frame


def mapped_price_target(forecast, prices, decision_close, annual_rf=.03,
                        valuations=None, error_scale=1., basis_shift=0.):
    """Condition a quarterly-average forecast on observations already available.

    For N quarter weekdays, observed sum A, n future weekdays and their ACT/365
    times h_j, solve T*N = A + S0*(n + mu*sum(h_j)). S0 is same-date NAV/ounces.
    Convert the expected bullion price at mean(h_j) to GLD using current ounces,
    fee decay, and the decision market close. Propagate forecast error by the
    exact derivative dQ/dT, rather than asserting a provider confidence score.
    """
    if forecast.get('mapping_type') != 'bullion_quarter_average':
        raise ValueError('A quarterly-average observation model is required.')
    decision = pd.Timestamp(decision_close)
    published = pd.Timestamp(forecast['published_on'])
    start, end = pd.Timestamp(forecast['period_start']), pd.Timestamp(forecast['period_end'])
    target, fee = forecast['target_usd_per_ounce'], forecast['annual_expense_ratio']
    forecast_error, basis_error = forecast['forecast_error_fraction'], forecast['basis_error_fraction']
    minimum = forecast['minimum_remaining_days']
    if (not published < start <= end or not isinstance(minimum, int) or minimum < 1
            or not np.isfinite([annual_rf,target,fee,forecast_error,basis_error,error_scale,basis_shift]).all()
            or annual_rf <= -1 or target <= 0 or not 0 <= fee < 1 or forecast_error <= 0
            or basis_error < 0 or error_scale <= 0 or abs(basis_shift) > .1):
        raise ValueError('Invalid forecast dates, fees or error assumptions.')
    if decision <= published or (end-decision).days < minimum:
        return None
    asset = forecast['asset']
    if (not isinstance(prices.index,pd.DatetimeIndex) or prices.index.has_duplicates
            or not prices.index.is_monotonic_increasing or asset not in prices or decision not in prices.index):
        raise ValueError('Exact decision price is required on a sorted unique index.')
    valuations = load_gold_valuations(forecast) if valuations is None else valuations
    known = valuations.loc[:decision]
    if (known.empty or decision not in known.index or known.index.has_duplicates
            or not known.index.is_monotonic_increasing or not np.isfinite(known.to_numpy()).all()
            or (known <= 0).any().any()):
        raise ValueError('Exact same-date issuer NAV and ounces are required; no stale decision pair.')
    ounces, nav = known.loc[decision, ['Ounces per share','NAV per share']].astype(float)
    current = float(prices.at[decision, asset])
    if not np.isfinite(current) or current <= 0:
        raise ValueError('Decision price must be finite and positive.')
    spot = nav / ounces
    quarter = pd.bdate_range(start, end)  # Explicit weekday averaging convention.
    observed, future = quarter[quarter <= decision], quarter[quarter > decision]
    bullion = known['NAV per share'] / known['Ounces per share']
    observed_values = bullion.reindex(bullion.index.union(observed)).ffill().reindex(observed)
    if observed_values.isna().any():
        raise ValueError('Missing already observed quarter valuations.')
    # Carry-forward is solely the declared holiday convention; never bridge an
    # absent price session that should have its own issuer valuation.
    required = prices.loc[start:decision].index.intersection(observed)
    if not required.isin(known.index).all():
        raise ValueError('Missing issuer valuation for an observed price session.')
    times = (future-decision).days.to_numpy(dtype=float)/365
    h = float(times.mean())
    slope = (target*len(quarter)-observed_values.sum()-len(future)*spot)/(spot*times.sum())
    future_spot = spot*(1+slope*h)
    if future_spot <= 0 or (1+slope*times).min() <= 0:
        raise ValueError('Quarterly forecast implies a nonpositive linear price path.')
    conversion = ounces*(1-fee)**h*(1+basis_shift)
    mapped = conversion*future_spot
    cash_horizon = (1+annual_rf)**h-1
    q = (mapped/current-1-cash_horizon)/h
    derivative = conversion/current*len(quarter)/times.sum()
    omega = (derivative*target*forecast_error*error_scale)**2 + (basis_error*error_scale/h)**2
    return {'view': (asset,None,float(q),1.), 'Omega annual variance': float(omega),
            'Asset':asset,'Published':str(published.date()),'Decision close':decision,
            'Mapping recorded':forecast['recorded_on'],'Quarter start':str(start.date()),
            'Quarter end':str(end.date()),'Quarter weekdays':len(quarter),
            'Observed quarter weekdays':len(observed),'Mean remaining years':h,
            'NAV per share':nav,'Ounces per share':ounces,'NAV-implied bullion':spot,
            'Decision ETF close':current,'Mapped mean future ETF price':mapped,
            'Implied annual bullion slope':float(slope),'Annual excess Q':float(q),
            'Forecast error fraction':forecast_error*error_scale,
            'Basis error fraction':basis_error*error_scale,'Basis shift assumption':basis_shift,
            'Evidence':'retrospective reconstruction; weekday average; NAV-implied bullion proxy'}


def stable_allocation(expected, covariance, current, cap=.35, risk_aversion=3.07,
                      max_active_weight=.15, transaction_cost_bps=10., holding_years=21/252):
    """Concave mean/variance utility minus explicit one-way trading costs.

    L1 epigraph variables make the cost term smooth for SLSQP. The static active
    bounds are relative to equal prior weights. Execution turnover is separately
    capped using actual holdings at the execution close, without refitting Q.
    """
    names = list(covariance.columns); n = len(names)
    mu, sigma, old = np.asarray(expected,float), covariance.to_numpy(float), np.asarray(current,float)
    if (list(covariance.index) != names or list(expected.index) != names
            or mu.shape != (n,) or old.shape != (n,) or not np.isfinite([*mu,*old,*sigma.ravel()]).all()
            or not np.allclose(sigma,sigma.T) or np.linalg.eigvalsh(sigma).min() < -1e-10
            or (old < -1e-10).any() or old.sum() > 1+1e-8
            or not np.isfinite([cap,risk_aversion,max_active_weight,transaction_cost_bps,holding_years]).all()
            or not 1/n <= cap <= 1 or risk_aversion <= 0 or max_active_weight < 0
            or transaction_cost_bps < 0 or holding_years <= 0):
        raise ValueError('Invalid optimizer labels, covariance, holdings or policy.')
    lower, upper = max(0.,1/n-max_active_weight), min(cap,1/n+max_active_weight)
    cost = transaction_cost_bps/10000/holding_years*.5
    def objective(x):
        w=x[:n]
        return risk_aversion/2*w@sigma@w-mu@w+cost*x[n:].sum()
    def gradient(x):
        return np.r_[risk_aversion*sigma@x[:n]-mu, np.full(n,cost)]
    # Initial cash movement is constant across fully invested candidate targets.
    constraints=[{'type':'eq','fun':lambda x:x[:n].sum()-1,
                  'jac':lambda x:np.r_[np.ones(n),np.zeros(n)]},
                 {'type':'ineq','fun':lambda x:x[n:]-(x[:n]-old),
                  'jac':lambda x:np.c_[-np.eye(n),np.eye(n)]},
                 {'type':'ineq','fun':lambda x:x[n:]+(x[:n]-old),
                  'jac':lambda x:np.c_[np.eye(n),np.eye(n)]}]
    x0=np.r_[np.full(n,1/n),np.abs(np.full(n,1/n)-old)]
    fit=minimize(objective,x0,jac=gradient,bounds=[(lower,upper)]*n+[(0,None)]*n,
                 constraints=constraints,method='SLSQP',options={'ftol':1e-12,'maxiter':1000})
    w=fit.x[:n]
    if (not fit.success or abs(w.sum()-1)>1e-7 or min(w)<lower-1e-7
            or max(w)>upper+1e-7 or (fit.x[n:]+1e-7<np.abs(w-old)).any()):
        raise RuntimeError(f'Cost-aware optimizer failed: {fit.message}')
    return pd.Series(w,index=names)


def limit_execution(target, current, max_turnover=.10, initial=False):
    """Interpolate toward the announced target to enforce actual one-way turnover.

    Includes cash. Initial deployment from all cash is explicitly exempt. Drift
    can leave an executed holding outside target bounds: report it, do not assert
    simultaneous hard bounds that can be infeasible under the turnover budget.
    """
    if (not target.index.equals(current.index) or not np.isfinite([*target,*current,max_turnover]).all()
            or not 0 < max_turnover <= 1 or min(target)<-1e-8 or min(current)<-1e-8
            or abs(target.sum()-1)>1e-7 or abs(current.sum()-1)>1e-7):
        raise ValueError('Execution needs matching long-only budgeted holdings and positive turnover cap.')
    turnover=float(np.abs(target-current).sum()/2)
    if initial and not (abs(current['Cash']-1)<1e-8):
        raise ValueError('Initial exception requires an all-cash portfolio.')
    fraction=1. if initial or turnover<=max_turnover else max_turnover/turnover
    return current+fraction*(target-current),fraction


def run_jpm_view_study(prices, prior_result, forecast, annual_rf=.03,
                       valuations=None, error_scale=1., basis_shift=0., policy=None,
                       prospective_after=None):
    """Dated signals and actual execution guard, with five common-cost controls.

    Forecast fit at t-2, execution at t-1, first earned return t. Current holdings
    in the objective stop at t-2. The turnover guard alone sees holdings at t-1.
    All paths start in cash and use identical execution dates and cash accounting.
    prospective_after additionally forbids fitting on or before the frozen date.
    """
    policy = dict(risk_aversion=3.07,max_active_weight=.15,max_one_way_turnover=.10,
                  holding_years=21/252,fixed_gold_overweight=.05,initial_cash_entry_exempt=True) if policy is None else dict(policy)
    if policy['initial_cash_entry_exempt'] is not True or not 0 <= policy['fixed_gold_overweight'] <= .15:
        raise ValueError('Invalid declared initial-entry or fixed-gold policy.')
    valuations = load_gold_valuations(forecast) if valuations is None else valuations
    all_returns=prior_result['returns']; dates=prior_result['portfolio_returns'].index
    expiry=pd.Timestamp(forecast['period_end'])-pd.Timedelta(days=forecast['minimum_remaining_days']-1)
    schedule=prior_result['view_history'].index
    possible=all_returns.index[all_returns.index>=expiry]
    if len(possible):
        loc=all_returns.index.get_loc(possible[0])
        if loc+1<len(all_returns) and all_returns.index[loc+1] in dates:
            schedule=schedule.union(all_returns.index[loc+1:loc+2])
    if prospective_after is not None:
        cutoff=max(pd.Timestamp(prospective_after),pd.Timestamp(forecast['recorded_on']))
        schedule=pd.DatetimeIndex([d for d in schedule if all_returns.index[all_returns.index.get_loc(d)-2]>cutoff])
        if not len(schedule):
            raise ValueError('No subsequent decision/execution/return observations after freeze.')
        dates=dates[dates>=schedule[0]]
    schedule=schedule.sort_values()
    labels=['JPM-mapped scenario','Matched prior','Fixed GLD overweight','GLD exposure matched','Risk matched equal weight']
    columns=ASSETS+['Cash']; equal=pd.Series(1/len(ASSETS),index=ASSETS)
    cash=prior_result['cash_returns'].loc[dates]
    tables={name:[] for name in labels}; states={name:pd.Series([0.]*len(ASSETS)+[1.],index=columns) for name in labels}
    audits=[]; control_audits=[]; previous_effective=None
    for effective in schedule:
        pos=all_returns.index.get_loc(effective)
        decision,execution=all_returns.index[pos-2],all_returns.index[pos-1]
        # Propagate previous post-trade holdings through only the decision close.
        for name in labels:
            if previous_effective is not None:
                for day in all_returns.loc[previous_effective:decision].index:
                    st=states[name]
                    _,w,c,_=portfolio_step(st[ASSETS].to_numpy(),st['Cash'],all_returns.loc[day,ASSETS].to_numpy(),float(prior_result['cash_returns'].at[day]),0.)
                    states[name]=pd.Series(np.r_[w,c],index=columns)
        train=all_returns.loc[:decision]
        if prior_result['training_window_days'] is not None:train=train.tail(prior_result['training_window_days'])
        covariance=estimate_covariance(train[ASSETS],prior_result['covariance_estimator'])
        mapping=mapped_price_target(forecast,prices,decision,annual_rf,valuations,error_scale,basis_shift)
        if execution>=expiry:mapping=None
        details=black_litterman(covariance,risk_aversion=policy['risk_aversion'],views=[] if mapping is None else [mapping['view']],
            omega=None if mapping is None else [[mapping['Omega annual variance']]],return_details=True)
        announced={}
        for name,expected in [('JPM-mapped scenario',details['posterior']),('Matched prior',details['prior'])]:
            w=stable_allocation(expected,covariance,states[name][ASSETS],prior_result['max_weight'],
                policy['risk_aversion'],policy['max_active_weight'],prior_result['transaction_cost_bps'],policy['holding_years'])
            announced[name]=pd.concat([w,pd.Series({'Cash':0.})])
        fixed=equal.copy()
        if mapping is not None:
            gold=min(equal['GLD']+policy['fixed_gold_overweight'],prior_result['max_weight'])
            fixed[:]=(1-gold)/(len(ASSETS)-1);fixed['GLD']=gold
        announced[labels[2]]=pd.concat([fixed,pd.Series({'Cash':0.})])
        # Only execution holdings, not Q/covariance, incorporate the next close.
        for name in labels:
            st=states[name]
            if previous_effective is not None:
                _,w,c,_=portfolio_step(st[ASSETS].to_numpy(),st['Cash'],all_returns.loc[execution,ASSETS].to_numpy(),float(prior_result['cash_returns'].at[execution]),0.)
                states[name]=pd.Series(np.r_[w,c],index=columns)
        jpm_executed,_=limit_execution(announced[labels[0]],states[labels[0]],policy['max_one_way_turnover'],previous_effective is None)
        gold=float(jpm_executed['GLD']);exposure=equal.copy();exposure[:]=(1-gold)/(len(ASSETS)-1);exposure['GLD']=gold
        announced[labels[3]]=pd.concat([exposure,pd.Series({'Cash':0.})])
        strategy_vol=float(np.sqrt(jpm_executed[ASSETS]@covariance@jpm_executed[ASSETS]))
        equal_vol=float(np.sqrt(equal@covariance@equal))
        scale=min(1.,strategy_vol/equal_vol) if equal_vol>0 else 1.
        announced[labels[4]]=pd.concat([equal*scale,pd.Series({'Cash':1-scale})])
        for name in labels:
            target,fraction=limit_execution(announced[name],states[name],policy['max_one_way_turnover'],previous_effective is None)
            actual_turnover=float(np.abs(target-states[name]).sum()/2)
            tables[name].append(target.rename(effective));states[name]=target
            control_audits.append({'Date':effective,'Strategy':name,'Execution fraction':fraction,'One-way turnover':actual_turnover,
                'Ex ante volatility':float(np.sqrt(target[ASSETS]@covariance@target[ASSETS])),
                'GLD weight':target['GLD'],'GLD difference vs strategy':target['GLD']-jpm_executed['GLD'],
                'Risk difference vs strategy':float(np.sqrt(target[ASSETS]@covariance@target[ASSETS]))-strategy_vol,
                'Executed max asset weight':target[ASSETS].max(),'Target upper bound':min(prior_result['max_weight'],1/len(ASSETS)+policy['max_active_weight'])})
        row={'Date':effective,'Decision close':decision,'Execution close':execution,'Active':mapping is not None,
             'Source publication':forecast['published_on'],'Mapping recorded':forecast['recorded_on'],
             'Evidence':'subsequent frozen-rule evaluation' if prospective_after is not None else 'retrospective publication reconstruction'}
        if mapping is not None:
            row.update({k:v for k,v in mapping.items() if k not in {'view','Evidence'}})
            row['Effective marginal weight']=float(details['views']['Effective marginal weight'].iloc[0])
        audits.append(row);previous_effective=effective
    paths,ledgers={},{};target_tables={k:pd.DataFrame(v) for k,v in tables.items()}
    for name,t in target_tables.items():
        paths[name],ledgers[name]=evaluate_targets(all_returns.loc[dates,ASSETS],t,annual_rf,
            prior_result['transaction_cost_bps'],prior_result['annual_short_borrow_bps'],cash_returns=cash)
    paths=pd.DataFrame(paths);audit=pd.DataFrame(audits).set_index('Date');active=audit.index[audit.Active]
    def summarize(period):
        if period.empty:return pd.DataFrame()
        frame=performance_table(period,annual_rf,cash_returns=cash.loc[period.index])
        frame['Total period return']=(1+period).prod()-1;frame['Sessions']=len(period)
        frame['First date']=str(period.index[0].date());frame['Last date']=str(period.index[-1].date())
        for name in frame.index:
            frame.loc[name,'One-way turnover']=ledgers[name].loc[period.index.intersection(ledgers[name].index),'One-way turnover'].sum()
        return frame
    last=prices.index[-1];mapping=mapped_price_target(forecast,prices,last,annual_rf,valuations,error_scale,basis_shift)
    train=all_returns.loc[:last]
    if prior_result['training_window_days'] is not None:train=train.tail(prior_result['training_window_days'])
    covariance=estimate_covariance(train[ASSETS],prior_result['covariance_estimator'])
    details=black_litterman(covariance,risk_aversion=policy['risk_aversion'],views=[] if mapping is None else [mapping['view']],
        omega=None if mapping is None else [[mapping['Omega annual variance']]],return_details=True)
    # Latest illustration references holdings drifted to the last observed close.
    latest={}
    for name,expected in [('JPM-mapped scenario',details['posterior']),('Matched prior',details['prior'])]:
        st=states[name]
        for day in all_returns.loc[previous_effective:last].index:
            _,w,c,_=portfolio_step(st[ASSETS].to_numpy(),st['Cash'],all_returns.loc[day,ASSETS].to_numpy(),float(prior_result['cash_returns'].at[day]),0.)
            st=pd.Series(np.r_[w,c],index=columns)
        latest[name]=stable_allocation(expected,covariance,st[ASSETS],prior_result['max_weight'],policy['risk_aversion'],
            policy['max_active_weight'],prior_result['transaction_cost_bps'],policy['holding_years'])
    latest=pd.DataFrame(latest);latest['Difference']=latest.iloc[:,0]-latest.iloc[:,1]
    return {'returns':paths,'targets':target_tables,'ledgers':ledgers,'audit':audit,
            'control_audit':pd.DataFrame(control_audits).set_index(['Date','Strategy']),
            'summary':summarize(paths),'since_first_activation':summarize(paths.loc[active[0]:]) if len(active) else pd.DataFrame(),
            'latest_weights':latest,'latest_details':details,'latest_mapping':mapping,'latest_price_date':last,
            'expiry_execution_boundary':expiry,'policy':policy}


def jpm_sensitivity(prices, prior_result, forecast, annual_rf=.03, policy=None):
    """Report all predeclared error/basis cases without performance selection."""
    valuations=load_gold_valuations(forecast)
    cases=[('base',1.,0.),('error_half',.5,0.),('error_double',2.,0.),('basis_minus_2pct',1.,-.02),('basis_plus_2pct',1.,.02)]
    rows,runs=[],{}
    for label,scale,basis in cases:
        run=run_jpm_view_study(prices,prior_result,forecast,annual_rf,valuations,scale,basis,policy)
        runs[label]=run;s=run['since_first_activation']
        row={'Case':label,'Forecast error fraction':forecast['forecast_error_fraction']*scale,
             'Basis error fraction':forecast['basis_error_fraction']*scale,'Basis shift':basis,
             'Active rebalances':int(run['audit'].Active.sum()),'Status':'evaluated reconstruction' if len(s) else 'no eligible dates'}
        if len(s):
            row.update({'First date':s.iloc[0]['First date'],'Last date':s.iloc[0]['Last date'],'Sessions':int(s.iloc[0]['Sessions']),
                'Scenario period return':s.loc['JPM-mapped scenario','Total period return'],
                'Prior period return':s.loc['Matched prior','Total period return'],
                'Period return difference':s.loc['JPM-mapped scenario','Total period return']-s.loc['Matched prior','Total period return'],
                'Scenario maximum drawdown':s.loc['JPM-mapped scenario','Max drawdown']})
        rows.append(row)
    return pd.DataFrame(rows).set_index('Case'),runs
````

### Calculation: Five-factor time-series regression

````python
def factor_regression(monthly_asset_returns, monthly_factors, hac_lags=3):
    """OLS excess-return regressions on the Fama-French five factors.

    Both inputs must already be decimals. The output reports annualized alpha,
    factor loadings, IID t-statistics, Bartlett HAC uncertainty with Holm-adjusted
    alpha p-values, R-squared and observation count.
    """
    assets = _numeric_frame(monthly_asset_returns, 'monthly_asset_returns')
    factors = _numeric_frame(monthly_factors, 'monthly_factors')
    required = FACTOR_COLUMNS + ['RF']
    if not set(required).issubset(factors.columns):
        raise ValueError(f'monthly_factors requires {required}.')
    joined = assets.join(factors[required], how='inner').dropna()
    if len(joined) < 24:
        raise ValueError('Too few aligned months for factor regression; at least 24 required.')
    if not isinstance(hac_lags, int) or not 0 <= hac_lags < len(joined):
        raise ValueError('HAC lags must be a nonnegative integer below the sample size.')
    if not joined.index.to_period('M').equals(pd.period_range(joined.index[0], joined.index[-1], freq='M')):
        raise ValueError('HAC attribution requires contiguous monthly observations.')
    import statsmodels.api as sm
    from statsmodels.stats.sandwich_covariance import cov_hac
    from scipy.stats import norm
    x = np.column_stack([np.ones(len(joined)), joined[FACTOR_COLUMNS].to_numpy()])
    if np.linalg.matrix_rank(x) < x.shape[1]:
        raise ValueError('Factor design matrix is rank deficient.')
    xtx_inverse = np.linalg.pinv(x.T @ x)
    rows = []
    for asset in assets.columns:
        y = (joined[asset] - joined['RF']).to_numpy()
        coefficients = np.linalg.lstsq(x, y, rcond=None)[0]
        robust_covariance = cov_hac(sm.OLS(y, x).fit(), nlags=hac_lags, use_correction=True)
        robust_se = np.sqrt(np.maximum(np.diag(robust_covariance), 0))
        robust_z = np.divide(coefficients, robust_se, out=np.full_like(coefficients, np.nan), where=robust_se > 1e-14)
        residuals = y - x @ coefficients
        dof = len(y) - x.shape[1]
        residual_variance = float(residuals @ residuals / dof)
        standard_errors = np.sqrt(np.diag(xtx_inverse) * residual_variance)
        t_stats = np.divide(coefficients, standard_errors,
                            out=np.full_like(coefficients, np.nan), where=standard_errors > 0)
        total = float(((y - y.mean()) ** 2).sum())
        r_squared = 1 - float(residuals @ residuals) / total if total > 0 else np.nan
        row = {'Portfolio': asset, 'Annual alpha': coefficients[0] * 12,
               'Alpha t-stat': t_stats[0], 'R-squared': r_squared,
               'Observations': len(y),
               'First month': str(joined.index.min().date()),
               'Last month': str(joined.index.max().date())}
        row.update({'Annual alpha HAC SE': robust_se[0] * 12,
                    'Alpha HAC z': robust_z[0], 'Alpha HAC p': 2 * norm.sf(abs(robust_z[0])),
                    'Alpha HAC 95% lower': (coefficients[0] - 1.96 * robust_se[0]) * 12,
                    'Alpha HAC 95% upper': (coefficients[0] + 1.96 * robust_se[0]) * 12,
                    'HAC lags': hac_lags})
        for number, factor in enumerate(FACTOR_COLUMNS, start=1):
            row[f'{factor} loading'] = coefficients[number]
            row[f'{factor} t-stat'] = t_stats[number]
            row[f'{factor} HAC SE'] = robust_se[number]
            row[f'{factor} HAC z'] = robust_z[number]
        rows.append(row)
    table = pd.DataFrame(rows).set_index('Portfolio')
    # Holm controls the alpha-test family across displayed portfolios, not all research choices.
    from statsmodels.stats.multitest import multipletests
    valid = table['Alpha HAC p'].notna()
    table['Alpha HAC Holm p'] = np.nan
    if valid.any():
        table.loc[valid, 'Alpha HAC Holm p'] = multipletests(table.loc[valid, 'Alpha HAC p'], method='holm')[1]
    return table
````

### Calculation: Full-model fitted excess returns

````python
def factor_fit_diagnostics(monthly_returns, factors):
    """Fitted excess returns use every factor, not a partial market-only line."""
    coefficients = factor_regression(monthly_returns, factors)
    dates = monthly_returns.index.intersection(factors.index)
    actual = monthly_returns.loc[dates].sub(factors.loc[dates, 'RF'], axis=0)
    fitted = pd.DataFrame(index=dates)
    for name in monthly_returns:
        beta = coefficients.loc[name, [f'{f} loading' for f in FACTOR_COLUMNS]].to_numpy(dtype=float)
        fitted[name] = coefficients.loc[name, 'Annual alpha'] / 12 + factors.loc[dates, FACTOR_COLUMNS].to_numpy() @ beta
    return coefficients, actual, fitted
````

### Calculation: Proportional allocation with repeated cap enforcement

````python
def capped_proportional_weights(scores, cap=0.35):
    """Allocate proportionally, repeatedly fixing saturated assets at their cap."""
    scores = pd.Series(scores, dtype=float).copy()
    if (scores.index.has_duplicates or not np.isfinite(scores).all()
            or (scores < 0).any() or scores.sum() <= 0
            or not 1 / len(scores) <= cap <= 1):
        raise ValueError('Scores must be nonnegative with a feasible cap and positive total.')
    weights = pd.Series(0.0, index=scores.index)
    free = list(scores.index)
    remaining = 1.0
    while free:
        values = scores.loc[free]
        proposal = remaining * (values / values.sum() if values.sum() > 0 else pd.Series(1 / len(free), index=free))
        saturated = proposal.index[proposal > cap + 1e-12]
        if len(saturated) == 0:
            weights.loc[free] = proposal
            break
        weights.loc[saturated] = cap
        remaining = 1 - weights.sum()
        free = [name for name in free if name not in saturated]
    return weights
````

### Calculation: Factor-threshold sensitivity rule

````python
def factor_tilt(weights, loadings, positive_threshold, negative_threshold, cap=.35):
    """Original +/-10% per-factor heuristic, with repeated cap enforcement."""
    base = pd.Series(weights, dtype=float)
    loadings = pd.DataFrame(loadings).loc[base.index]
    if (not np.isfinite(loadings.to_numpy()).all() or positive_threshold <= 0
            or negative_threshold <= 0):
        raise ValueError('Loadings must be finite and thresholds positive.')
    positive = (loadings >= positive_threshold).sum(axis=1)
    negative = (loadings <= -negative_threshold).sum(axis=1)
    scores = base * 1.1 ** positive * .9 ** negative
    return capped_proportional_weights(scores, cap)
````

### B. Factor-threshold allocation heuristic

The original rule multiplies an ETF weight by 1.1 for a positive loading and 0.9
for a negative loading exceeding a threshold. A loading's sign alone is not a
forecast of a factor premium. This is a sensitivity experiment, not Fama–MacBeth
estimation or a validated investment rule.

We restore both the original numerical thresholds (+0.010 / -0.008) and an adaptive
absolute-loading mean-plus-standard-deviation threshold. The old thresholds were
chosen using inconsistent units; here they are explicitly **illustrative** on
corrected dimensionless loadings. All five factors contribute; output is capped
and normalized after the multipliers. Fits stop before the last rebalance and
use the available factor vintage. Before/after monthly statistics use 12 periods
per year and the same calibration sample. They are gross in-sample constant-weight
illustrations, not the net walk-forward strategy results.

### Calculation: Named macro series and descriptive normalization

````python
def load_inflation(path=None):
    """Global annual CPI rates in percentage points, selected by Series Name."""
    path = Path(path) if path else DATA_DIR / 'Inflation-data.xlsx'
    raw = pd.read_excel(path, sheet_name='Aggregate', engine='openpyxl')
    if 'Series Name' not in raw:
        raise ValueError('Inflation Aggregate sheet requires Series Name.')
    selected = raw.loc[raw['Series Name'].isin(INFLATION_SERIES)]
    if len(selected) != 3 or selected['Series Name'].duplicated().any():
        raise ValueError('Inflation file must contain each global CPI series exactly once.')
    years = [c for c in raw if re.fullmatch(r'\d{4}', str(c))]
    if not years:
        raise ValueError('Inflation file has no annual observation columns.')
    result = selected.set_index('Series Name')[years].T.rename(columns=INFLATION_SERIES)
    result.index = pd.Index([int(y) for y in years], name='Year')
    result = result.apply(pd.to_numeric, errors='raise').sort_index()
    if np.isinf(result.to_numpy()).any():
        raise ValueError('Inflation contains infinite values.')
    result.attrs.update(unit='percent per year', sha256=file_sha256(path))
    return result


def load_liquidity(path=None):
    """BIS bank credit as percent of GDP, selected by its full economic key."""
    path = Path(path) if path else DATA_DIR / 'WS_GLI_csv_col.csv'
    raw = pd.read_csv(path, dtype=str)
    missing = set(BIS_SERIES) - set(raw.columns)
    if missing:
        raise ValueError(f'BIS file lacks series metadata: {sorted(missing)}')
    mask = pd.Series(True, index=raw.index)
    for key, value in BIS_SERIES.items():
        mask &= raw[key].str.strip().eq(value)
    selected = raw.loc[mask]
    if len(selected) != 1:
        raise ValueError(f'BIS economic key must match exactly one series, found {len(selected)}.')
    quarters = [c for c in raw if re.fullmatch(r'\d{4}-Q[1-4]', c)]
    if not quarters:
        raise ValueError('BIS file has no quarterly observation columns.')
    values = pd.to_numeric(selected.iloc[0][quarters], errors='raise')
    result = pd.Series(values.to_numpy(dtype=float),
                       index=pd.PeriodIndex(quarters, freq='Q'), name='Bank credit (% of GDP)').sort_index()
    if np.isinf(result.to_numpy()).any() or result.notna().sum() == 0:
        raise ValueError('BIS observations must contain finite data.')
    result.index.name = 'Quarter'
    result.attrs.update(unit='percent of GDP', series_key=BIS_SERIES, sha256=file_sha256(path))
    return result


def normalize_context(series):
    """Full-sample z scores for display only; never a point-in-time signal."""
    values = pd.Series(series, dtype=float).dropna()
    if len(values) < 2 or not np.isfinite(values).all() or values.std() <= 0:
        raise ValueError('Context normalization needs finite, nonconstant observations.')
    return (values - values.mean()) / values.std()
````

## Implemented remaining improvements — separate October 16 amendment

The new functions and calculations below preserve all frozen baseline cells. Common risk is tested retrospectively; calibration is gated by completed comparable outcomes; the paper journal is read without creating or changing decisions.

````python
"""Separate October 16 research extensions. No changes to the frozen model.

All decisions are paper research. Functions are also embedded in the amended
notebook; no broker, scheduler, network, or production imports occur here.
"""
import hashlib
import json
from datetime import datetime, timezone
from pathlib import Path
import numpy as np
import pandas as pd
from scipy.optimize import linprog
import pandas_market_calendars as mcal


def feasible_cash_scale(direction, current, cap=.275, turnover_limit=.10, initial=False):
    """Exact linear-program interval on a fixed risky allocation plus cash.

    Turnover includes cash; long-only and unlevered. Empty intervals are exposed.
    This chooses exposure, never a new return forecast or asset direction.
    """
    d=np.asarray(direction,dtype=float);old=np.asarray(current,dtype=float);n=len(d)
    if (old.shape!=(n+1,) or not np.isfinite(np.r_[d,old,cap,turnover_limit]).all()
            or (d<0).any() or (old < -1e-10).any() or abs(d.sum()-1)>1e-8
            or abs(old.sum()-1)>1e-8 or not 0<cap<=1 or not 0<turnover_limit<=1):
        raise ValueError('Invalid cash-scale direction, holdings or constraints.')
    if initial and not np.allclose(old,np.r_[np.zeros(n),1.]):
        raise ValueError('Initial deployment exception requires all cash.')
    # w(s) = [s*d, 1-s]. Epigraph t >= abs(w(s)-old).
    slope=np.r_[d,-1.];intercept=np.r_[np.zeros(n),1.]-old
    a=np.vstack([np.c_[slope,-np.eye(n+1)],np.c_[-slope,-np.eye(n+1)],np.r_[0.,np.ones(n+1)]])
    b=np.r_[-intercept,intercept,2*(1. if initial else turnover_limit)]
    bound=min(1.,cap/d.max());bounds=[(0.,bound)]+[(0,None)]*(n+1)
    low=linprog(np.r_[1.,np.zeros(n+1)],A_ub=a,b_ub=b,bounds=bounds,method='highs')
    high=linprog(np.r_[-1.,np.zeros(n+1)],A_ub=a,b_ub=b,bounds=bounds,method='highs')
    if low.status==2 or high.status==2:return None
    if not low.success or not high.success:raise RuntimeError('Cash-scale LP failed numerically.')
    return max(0.,float(low.x[0])),min(bound,float(high.x[0]))


def common_risk_pair(directions, currents, covariance, cap=.275, turnover_limit=.10, initial=False):
    """Intersect attainable volatility intervals, using the SAME prior covariance.

    Choose the highest shared risk no greater than either fully invested sleeve.
    If infeasible, both controls hold; the mismatch remains visible in the audit.
    Equality holds at rebalance under this covariance, not continuously or ex post.
    """
    names=list(covariance.columns);sigma=covariance.to_numpy(float)
    if (list(covariance.index)!=names or not np.isfinite(sigma).all()
            or not np.allclose(sigma,sigma.T) or np.linalg.eigvalsh(sigma).min()<=0
            or set(directions)!=set(currents) or len(directions)!=2):
        raise ValueError('Two controls and a positive definite labelled covariance required.')
    intervals={};vols={}
    for label,direction in directions.items():
        if list(direction.index)!=names or list(currents[label].index)!=names+['Cash']:
            raise ValueError('Asset order mismatch.')
        vols[label]=float(np.sqrt(direction@sigma@direction))
        intervals[label]=feasible_cash_scale(direction,currents[label],cap,turnover_limit,initial)
    feasible=all(x is not None for x in intervals.values())
    lower=max(intervals[k][0]*vols[k] for k in intervals) if feasible else None
    upper=min(intervals[k][1]*vols[k] for k in intervals) if feasible else None
    feasible=bool(feasible and lower<=upper+1e-10)
    targets={};rows=[]
    for label,d in directions.items():
        scale=float(upper/vols[label]) if feasible else None
        target=pd.Series(np.r_[scale*d,1-scale],index=names+['Cash']) if feasible else currents[label].copy()
        risk=float(np.sqrt(target[names]@sigma@target[names]));turnover=float(abs(target-currents[label]).sum()/2)
        if feasible and (turnover>(1 if initial else turnover_limit)+1e-7 or target[names].max()>cap+1e-7
                         or target.min() < -1e-8 or abs(target.sum()-1)>1e-8 or abs(risk-upper)>1e-8):
            raise RuntimeError('Common-risk execution failed independent constraint check.')
        targets[label]=target
        rows.append({'Strategy':label,'Status':'MATCHED' if feasible else 'INFEASIBLE_HOLD_BOTH',
            'Common volatility':upper if feasible else None,'Actual ex ante volatility':risk,'Risk gap':risk-upper if feasible else None,
            'Feasible minimum':None if intervals[label] is None else intervals[label][0]*vols[label],
            'Feasible maximum':None if intervals[label] is None else intervals[label][1]*vols[label],
            'Risky scale':scale,'One-way turnover':turnover,'Cash':target['Cash'],'Initial exception':initial})
    return targets,pd.DataFrame(rows).set_index('Strategy')


def replay_common_risk(prices, cash_returns, forecast, policy, schedule, annual_rf=.03, training_rows=252):
    """New retrospective paired experiment with an independent holdings path.

    Fit t-2. Freeze risky directions/covariance then; at t-1 only apply the
    predeclared cash-scaling rule using actual pretrade holdings. Earn at t.
    """
    returns=prices[ASSETS].pct_change(fill_method=None).iloc[1:]
    if not returns.index.equals(cash_returns.index):raise ValueError('Cash/date mismatch.')
    dates=returns.index[returns.index>=schedule[0]];labels=['Scenario common risk','Equal weight common risk']
    states={k:pd.Series(np.r_[np.zeros(len(ASSETS)),1.],index=ASSETS+['Cash']) for k in labels}
    tables={k:[] for k in labels};audits=[];previous=None
    equal=pd.Series(1/len(ASSETS),index=ASSETS)
    expiry=pd.Timestamp(forecast['period_end'])-pd.Timedelta(days=forecast['minimum_remaining_days']-1)
    schedule=pd.DatetimeIndex(schedule).sort_values()
    expiry_positions=np.flatnonzero(returns.index>=expiry)
    if len(expiry_positions) and expiry_positions[0]+1<len(returns):
        schedule=schedule.union(returns.index[expiry_positions[0]+1:expiry_positions[0]+2])
    for effective in schedule:
        pos=returns.index.get_loc(effective)
        if pos<2:raise ValueError('Need decision and execution observations before return.')
        decision,execution=returns.index[pos-2:pos]
        for label in labels:
            if previous is not None:
                for day in returns.loc[previous:decision].index:
                    st=states[label];_,w,c,_=portfolio_step(st[ASSETS],st['Cash'],returns.loc[day],cash_returns.at[day],0.)
                    states[label]=pd.Series(np.r_[w,c],index=ASSETS+['Cash'])
        covariance=estimate_covariance(returns.loc[:decision].tail(training_rows),'sample')
        mapping=mapped_price_target(forecast,prices,decision,annual_rf)
        if execution>=expiry:mapping=None
        detail=black_litterman(covariance,risk_aversion=policy['risk_aversion'],
            views=[] if mapping is None else [mapping['view']],omega=None if mapping is None else [[mapping['Omega annual variance']]],return_details=True)
        direction=stable_allocation(detail['posterior'],covariance,states[labels[0]][ASSETS],.35,
            policy['risk_aversion'],policy['max_active_weight'],10.,policy['holding_years'])
        directions={labels[0]:direction,labels[1]:equal}
        for label in labels:
            if previous is not None:
                st=states[label];_,w,c,_=portfolio_step(st[ASSETS],st['Cash'],returns.loc[execution],cash_returns.at[execution],0.)
                states[label]=pd.Series(np.r_[w,c],index=ASSETS+['Cash'])
        targets,audit=common_risk_pair(directions,states,covariance,
            cap=min(.35,1/len(ASSETS)+policy['max_active_weight']),turnover_limit=policy['max_one_way_turnover'],initial=previous is None)
        audit['First return']=effective;audit['Decision close']=decision;audit['Execution close']=execution
        audit['Fit last observation']=returns.loc[:decision].index[-1];audit['View active']=mapping is not None
        audits.append(audit.reset_index())
        for label in labels:tables[label].append(targets[label].rename(effective));states[label]=targets[label]
        previous=effective
    target_tables={k:pd.DataFrame(v) for k,v in tables.items()};paths={};ledgers={}
    for label in labels:
        paths[label],ledgers[label]=evaluate_targets(returns.loc[dates],target_tables[label],annual_rf,10.,100.,cash_returns.loc[dates])
    # Independent accounting replay must agree with the execution audit.
    audit=pd.concat(audits,ignore_index=True)
    for label in labels:
        actual=ledgers[label]['One-way turnover'].to_numpy()
        expected=audit.loc[audit.Strategy==label,'One-way turnover'].to_numpy()
        if not np.allclose(actual,expected,atol=1e-9):raise RuntimeError('Independent turnover ledger disagrees.')
    return {'returns':pd.DataFrame(paths),'targets':target_tables,'audit':audit,'ledgers':ledgers}


def basis_walkforward(prices, valuations, minimum=40, prior_sigma=.02, prior_strength=40):
    """Prior-only diagnostic of close/NAV timing basis; NOT forecast Omega.

    Zero-centred RMSE shrunk toward the existing assumption; no P&L tuning.
    NAV and ETF close differ in observation time. Serial dependence is retained.
    """
    if minimum<2 or prior_sigma<=0 or prior_strength<=0:raise ValueError('Invalid basis policy.')
    joined=pd.concat([prices['GLD'].rename('Close'),valuations['NAV per share'].rename('NAV')],axis=1).dropna()
    if (joined<=0).any().any() or not np.isfinite(joined).all().all():raise ValueError('Invalid basis inputs.')
    basis=joined.Close/joined.NAV-1;rows=[]
    for i in range(minimum,len(basis)):
        train=basis.iloc[:i];variance=(float((train**2).sum())+prior_strength*prior_sigma**2)/(len(train)+prior_strength)
        sigma=float(np.sqrt(variance));error=float(basis.iloc[i])
        rows.append({'Date':basis.index[i],'Training end':train.index[-1],'Training observations':len(train),
            'Past mean basis':float(train.mean()),'Past RMSE':float(np.sqrt((train**2).mean())),
            'Shrunk sigma':sigma,'Observed basis':error,'Within assumed 95pct band':abs(error)<=1.96*sigma})
    return basis.rename('Close/NAV minus one'),pd.DataFrame(rows)


def forecast_calibration(records, asof, minimum_cycles=8, prior_sigma=.10, prior_strength=8):
    """Use only completed, available, comparable, distinct forecast quarters.

    Same-quarter revisions are not independent samples. Eligibility requires
    matching provider/outcome definition and pre-quarter publication.
    """
    now=pd.Timestamp(asof)
    if now.tzinfo is None:raise ValueError('Aware calibration asof required.')
    if minimum_cycles<2 or prior_sigma<=0 or prior_strength<=0:raise ValueError('Invalid calibration policy.')
    errors=[];seen=set();audit=[]
    for record in records:
        r=dict(record);start=pd.Timestamp(r['period_start']);end=pd.Timestamp(r['period_end']);reason=[]
        available=pd.Timestamp(r['available_at_utc'])
        if available.tzinfo is None:raise ValueError('Aware outcome availability required.')
        if available>now:reason.append('not available at decision')
        if end>=now.tz_convert('America/New_York').tz_localize(None).normalize():reason.append('quarter unfinished')
        if pd.Timestamp(r['published_on'])>=start:reason.append('different forecast lead horizon')
        if not r['outcome_definition_verified']:reason.append('outcome proxy/calendar not verified')
        actual=r.get('actual');target=float(r['target'])
        if target<=0 or not np.isfinite(target):raise ValueError('Invalid forecast target.')
        if actual is None or not np.isfinite(actual) or actual<=0:reason.append('completed outcome missing')
        cycle=(r['period_start'],r['period_end'])
        if cycle in seen:reason.append('duplicate forecast quarter')
        if not reason:seen.add(cycle);errors.append(float(actual/target-1))
        audit.append(dict(id=r['id'],eligible=not reason,reason='; '.join(reason) or 'eligible',
            diagnostic_proxy_error=None if actual is None else float(actual/target-1)))
    enough=len(errors)>=minimum_cycles
    sigma=float(np.sqrt((sum(e*e for e in errors)+prior_strength*prior_sigma**2)/(len(errors)+prior_strength))) if enough else prior_sigma
    return {'eligible_cycles':len(errors),'minimum_cycles':minimum_cycles,'forecast_sigma':sigma,
            'status':'PRIOR_ONLY_SHRINKAGE_ESTIMATE' if enough else 'RETAIN_UNCALIBRATED_DEFAULT','audit':audit}


def horizon_status(forecast, recorded_at, sessions_observed=0, completed_cycles=0):
    """Separate quarter scoring, active-view attribution and strategy inference."""
    now=pd.Timestamp(recorded_at)
    if now.tzinfo is None:raise ValueError('Aware protocol timestamp required.')
    end=pd.Timestamp(forecast['period_end']);expiry=end-pd.Timedelta(days=forecast['minimum_remaining_days']-1)
    schedule=mcal.get_calendar('NYSE').schedule(start_date=now.tz_convert('America/New_York').date(),end_date=end+pd.Timedelta(days=10))
    future=schedule.loc[schedule.market_close>now]
    if len(future)<2:raise ValueError('No future execution and return sessions.')
    execution=future.iloc[0].market_close;earned=future.iloc[1].market_close
    active=int(((future.index[:-1]<expiry)&(future.index[1:]<=end)).sum())
    return {'recorded_at_utc':now.tz_convert('UTC').isoformat(),'next_execution_close_utc':execution.isoformat(),
        'first_return_close_utc':earned.isoformat(),'view_expiry_execution_boundary':str(expiry.date()),
        'remaining_active_return_sessions':active,'minimum_strategy_sessions':60,'minimum_completed_forecast_cycles':8,
        'strategy_inference_admitted':sessions_observed>=60 and completed_cycles>=8,
        'completed_quarter_score_admitted':now.tz_convert('America/New_York').tz_localize(None).normalize()>end,
        'score_rule':'Quarter average forecast error separately; portfolio returns across frozen cycles. No inference from this unfinished quarter.'}


def canonical_hash(document):
    return hashlib.sha256(json.dumps(document,sort_keys=True,separators=(',',':'),allow_nan=False).encode()).hexdigest()


def verify_decision_journal(directory):
    """Verify sequence and hash chain. Local tamper evidence, not external notarization."""
    entries=[];previous=None
    for number,path in enumerate(sorted(Path(directory).glob('*.json')),1):
        if path.name!=f'{number:06d}.json':raise ValueError('Missing/reordered journal entry.')
        doc=json.loads(path.read_text(encoding='utf-8'));payload={k:v for k,v in doc.items() if k!='sha256'}
        if doc['sha256']!=canonical_hash(payload) or doc['previous_sha256']!=previous:raise ValueError('Journal chain/tamper failure.')
        if entries and pd.Timestamp(doc['recorded_at_utc'])<pd.Timestamp(entries[-1]['recorded_at_utc']):raise ValueError('Journal time reversal.')
        entries.append(doc);previous=doc['sha256']
    return entries


def append_paper_decision(directory, payload):
    """Write once, before execution. No caller-supplied timestamp/backdating.

    A paper decision is NOT an execution receipt. Outcomes are admitted later
    through a separate pure evaluator, never by editing this entry.
    """
    folder=Path(directory);folder.mkdir(parents=True,exist_ok=True)
    entries=verify_decision_journal(folder);now=pd.Timestamp.now(tz='UTC')
    for key in ['sha256','previous_sha256','recorded_at_utc','sequence']:
        if key in payload:raise ValueError('Caller cannot supply journal identity/time fields.')
    if any(e['decision_id']==payload['decision_id'] for e in entries):raise ValueError('Duplicate decision ID.')
    available=pd.Timestamp(payload['inputs_available_at_utc']);execution=pd.Timestamp(payload['execution_close_utc']);earned=pd.Timestamp(payload['first_return_close_utc'])
    if any(t.tzinfo is None for t in [available,execution,earned]) or not available<=now<execution<earned:
        raise ValueError('Decision must follow input availability and precede execution/outcome.')
    if payload['solver_status']!='MATCHED':raise ValueError('Journal requires a feasible declared paired allocation.')
    if not np.isfinite(payload['transaction_cost_bps']) or not 0<=payload['transaction_cost_bps']<10000:
        raise ValueError('Invalid paper cost assumption.')
    for weights in payload['targets'].values():
        values=np.array(list(weights.values()),float)
        if 'Cash' not in weights or not np.isfinite(values).all() or min(values)<-1e-8 or abs(values.sum()-1)>1e-8:
            raise ValueError('Invalid journal target budget.')
    document=dict(payload,sequence=len(entries)+1,recorded_at_utc=now.isoformat(),previous_sha256=entries[-1]['sha256'] if entries else None)
    document['sha256']=canonical_hash(document)
    # Exclusive creation makes a concurrent sequence collision fail explicitly.
    with (folder/f'{len(entries)+1:06d}.json').open('x',encoding='utf-8') as f:json.dump(document,f,indent=2,allow_nan=False)
    return document


def evaluate_paper_interval(entry, closing_prices, cash_return, now=None):
    """Evaluate only the first fully observed holding interval, independently.

    Supplied close dates must exactly match the recorded interval. First-entry
    cost uses the declared all-cash starting state. No broker execution implied.
    """
    current=pd.Timestamp.now(tz='UTC') if now is None else pd.Timestamp(now)
    if entry['sha256']!=canonical_hash({k:v for k,v in entry.items() if k!='sha256'}):raise ValueError('Decision receipt was changed.')
    start=pd.Timestamp(entry['execution_close_utc']);end=pd.Timestamp(entry['first_return_close_utc'])
    if current.tzinfo is None or current<end:raise ValueError('Outcome has not completed.')
    if entry['starting_state']!='ALL_CASH_NEW_PAPER_ACCOUNT':raise ValueError('Unsupported initial state.')
    if list(closing_prices.index)!=[start.tz_convert('America/New_York').tz_localize(None).normalize(),end.tz_convert('America/New_York').tz_localize(None).normalize()]:
        raise ValueError('Exact recorded execution/return closes required.')
    if (closing_prices<=0).any().any() or not np.isfinite(closing_prices).all().all() or not np.isfinite(cash_return) or cash_return<=-1:
        raise ValueError('Invalid realized inputs.')
    r=closing_prices.iloc[1]/closing_prices.iloc[0]-1;rows=[]
    for label,target in entry['targets'].items():
        w=pd.Series(target);assets=w.drop('Cash')
        if list(assets.index)!=list(closing_prices.columns):raise ValueError('Outcome asset mismatch.')
        cost=(1-w['Cash'])*entry['transaction_cost_bps']/10000
        net=(1-cost)*(1+float(assets@r)+w['Cash']*cash_return)-1
        rows.append({'Strategy':label,'Net interval return':net,'Initial cost':cost,'Decision sha256':entry['sha256'],
                     'Evidence':'Recorded paper target with subsequently supplied closes; not an execution receipt'})
    return pd.DataFrame(rows)
````

## Changes implemented

1. **Decision capture and outcome separation.** `paper_research.py` verifies the new protocol, source/runtime/data hashes and latest completed NYSE close, then records a research allocation before the next execution close. Exclusive numbered files and a hash chain detect accidental modification or missing entries. Capture time comes from the wall clock; callers cannot backdate it. Its subsequent return is evaluated only in a separate function/command after the recorded interval completes; no outcome is invented now. A complete rewrite or tail deletion requires comparing with the recorded external head/hash; this limitation is explicit.

2. **Horizon-aware rules.** Quarter-average forecast error is scored only after quarter completion and outcome availability. Longer-run strategy inference requires at least 60 genuinely subsequent sessions and eight completed comparable forecast cycles in the new design. Those are declared research minimums, not a guarantee of statistical power. For a paper entry captured before the October 16 close, only **32 active return sessions** remain before the existing December 2 execution-expiry boundary. Calendar calculation handles weekends, holidays and early closes. The 60-session rule is not lowered to make this quarter pass. A qualified cycle registry and continued recording are still required before either gate can pass.

3. **A common-risk control that actually matches.** Both risky portfolio directions are determined using data through decision close t-2 and the same trailing 252-session annual covariance. At execution close t-1, a predeclared scalar cash overlay uses the then-current holdings and frozen covariance/directions. Linear programs find each sleeve's feasible exposure interval under the same 27.5% asset cap, 10% one-way turnover limit including cash, no leverage and an explicit initial cash-entry exception. The overlap defines the highest jointly feasible volatility. If there is no overlap, both hold and the audit exposes the mismatch. No expected return is refitted at execution. Equal risk is a rebalance property under the prior covariance, not guaranteed realized or continuous risk equality.

4. **Prior-only uncertainty diagnostics and admission gates.** A forecast-error estimator accepts only completed, available, distinct quarters with comparable forecast lead horizons and verified outcome definitions. It shrinks mean squared forecast errors toward the existing 10% assumption after the declared minimum of eight qualifying quarters; below that, the assumption is retained and labelled uncalibrated. Daily close/NAV basis diagnostics use only prior observations, with 40 initial pairs and a 40-pair variance prior at 2%. The empirical basis estimate is not inserted into Omega because daily timing basis does not identify the full remaining-quarter conversion error.

5. **Traceability.** The amendment design is recorded before the new retrospective comparison. The paper protocol is separately frozen after validation, with the original protocol as a dependency. Notebook source hashes, test results, data identities, decision receipt and outputs are recorded. No baseline pin was relaxed, historical observations spliced, parameter selected for a better result, or trading integration activated.
